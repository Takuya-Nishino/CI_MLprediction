```python
# =========================
# Cell 1: 開発コード（フルコード：SHAP用の入力行列保存まで含む）
# =========================
# -*- coding: utf-8 -*-
"""
Stroke modeling pipeline (paper-ready, leak-safe)  [NO FIGURES]
Best-form codebase (fixed FS + consistent calibration)  [PATCHED: MI + CITL + cache signature + SHAP input save]

Leak-safety notes (CODE = source of truth):
- Feature selection is performed ONCE per task, using TRAIN ONLY.
- CV / preprocessing / MI are fit on fold-train only (OOF leak-safe).
- Platt calibration is fit on OOF predictions (train only).
- Temporal validation uses its OWN feature-selection cache (no reuse from holdout).

PATCH NOTES (2025-12-27):
1) FIX: LR+MI path previously had NO missing values because WinsorLogZ always median-imputed.
   -> Added WinsorLogZ(impute=True/False); LR+MI uses impute=False so IterativeImputer is effective.
2) FIX: CITL sklearn fallback was not offset-only (slope not fixed to 1).
   -> Implemented Newton solver for intercept-only CITL with fixed offset logit(p) (statsmodels optional).
3) ADD: FS cache signature (feat_cols + cfg) to avoid silent reuse when columns/cfg change.
4) ADD: SHAP-ready X matrices saved from the FINAL pipeline inputs to avoid RAW re-load / key mismatch.
   -> outputs to shap_data/ (parquet preferred; csv fallback)

Outputs:
- predictions/*.csv
- models/*.joblib
- shap_data/*.(parquet|csv)   <-- NEW
- Evaluation_*.xlsx
- best_params.json
- Split_Common.csv
- Split_TimeTemporal.csv
- env.txt
- config.json
"""

from __future__ import annotations

# ===================== THREAD CAPS =====================
import os as _os
_os.environ.setdefault("OMP_NUM_THREADS", "1")
_os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")
_os.environ.setdefault("MKL_NUM_THREADS", "1")
_os.environ.setdefault("VECLIB_MAXIMUM_THREADS", "1")
_os.environ.setdefault("NUMEXPR_NUM_THREADS", "1")
_os.environ.setdefault("CUDA_VISIBLE_DEVICES", "")

# ===================== IMPORTS =====================
import os, sys, json, gc, random, warnings, logging, platform, argparse, hashlib
from pathlib import Path
from typing import Dict, List, Tuple, Optional, Any

import numpy as np
import pandas as pd
warnings.filterwarnings("ignore")

import optuna
from optuna.samplers import TPESampler
from optuna.pruners import MedianPruner

from sklearn.model_selection import StratifiedShuffleSplit, StratifiedKFold, KFold, train_test_split
from sklearn.experimental import enable_iterative_imputer  # noqa
from sklearn.impute import IterativeImputer
from sklearn.metrics import roc_auc_score, average_precision_score, r2_score, median_absolute_error
from sklearn.linear_model import LogisticRegression
from sklearn.utils.class_weight import compute_class_weight

import lightgbm as lgb

try:
    import xgboost as xgb
    _HAS_XGB = True
except Exception:
    _HAS_XGB = False

_HAS_SM = True
try:
    import statsmodels.api as sm
    from statsmodels.stats.outliers_influence import variance_inflation_factor as _sm_vif
except Exception:
    _HAS_SM = False

try:
    import joblib
    _HAS_JOBLIB = True
except Exception:
    _HAS_JOBLIB = False

# ===================== CONFIG =====================
MODE = "MAX"  # "SPEED" | "MAX"
RANDOM_STATE = 42

RAW_PATH = r"C:\Users\tears\Desktop\Study\2025\12_CI\004_CI\Merge_20250924.xlsx"

TARGET_OUTCOME = "Events"     # 0/1
TARGET_LOS     = "在院日数"
INDEX_COL      = "INDEX"
IGNORE_COLS    = ["転帰"]     # optional leakage columns

LOS_CUTOFF_DAYS = 21
ENABLE_LOSBIN = True

RUN_DIR = Path(f"results_paper_ready_seed{RANDOM_STATE}")
RUN_DIR.mkdir(parents=True, exist_ok=True)
ART_DIR = RUN_DIR / "artifacts"; ART_DIR.mkdir(exist_ok=True)
PRED_DIR = RUN_DIR / "predictions"; PRED_DIR.mkdir(exist_ok=True)
MODEL_DIR = RUN_DIR / "models"; MODEL_DIR.mkdir(exist_ok=True)

# ---- NEW: SHAP-ready inputs ----
SHAP_DIR = RUN_DIR / "shap_data"; SHAP_DIR.mkdir(exist_ok=True)

EVAL_XLSX = RUN_DIR / "Evaluation_All.xlsx"
EVAL_TEMP_XLSX = RUN_DIR / "Evaluation_Temporal.xlsx"
BESTPARAMS_JSON = RUN_DIR / "best_params.json"

# --- IMPORTANT: separate FS cache files to avoid “temporal reusing holdout FS” ---
FEATURES_JSON_HOLDOUT = RUN_DIR / "selected_features_by_task_holdout.json"
FEATURES_JSON_TEMPORAL = RUN_DIR / "selected_features_by_task_temporal.json"

SPLIT_CSV = RUN_DIR / "Split_Common.csv"
SPLIT_TIME_CSV = RUN_DIR / "Split_TimeTemporal.csv"
ENV_TXT = RUN_DIR / "env.txt"
CFG_JSON = RUN_DIR / "config.json"

def _cfg_by_mode(mode: str):
    if mode.upper() == "MAX":
        return dict(
            # MI (LR only)
            MI_M=10,
            MI_MAX_ITER=25,
            MI_NEAREST=20,
            # CV / Optuna
            N_SPLITS_CLS=8,
            N_SPLITS_REG=8,
            TUNE_SUBSAMPLE_FRAC=0.8,
            N_TRIALS=120,
            TUNE_TIMEOUT_SEC=1800,
            # Feature selection (ONCE per task, train-only)
            TOPK_FEATURES=64,
            VIF_THR=8.0,
            VIF_NUMERIC_ONLY=True,         # safer: VIF on continuous-ish columns only
            PROTECTED_ALWAYS=["NIHSS total"],
            # Training
            EARLY_STOP=100,
            LGBM_N_EST_CLS=800,
            LGBM_N_EST_REG=800,
            XGB_N_EST_CLS=1200,
            XGB_N_EST_REG=1200,
            # Bootstrap (within-imputation variance)
            BOOT_B_CLASS=1000,
            BOOT_B_REG=1000,
        )
    return dict(
        MI_M=5, MI_MAX_ITER=20, MI_NEAREST=20,
        N_SPLITS_CLS=5, N_SPLITS_REG=5,
        TUNE_SUBSAMPLE_FRAC=0.6, N_TRIALS=30, TUNE_TIMEOUT_SEC=300,
        TOPK_FEATURES=48, VIF_THR=8.0, VIF_NUMERIC_ONLY=True, PROTECTED_ALWAYS=["NIHSS total"],
        EARLY_STOP=50,
        LGBM_N_EST_CLS=600, LGBM_N_EST_REG=600,
        XGB_N_EST_CLS=800, XGB_N_EST_REG=800,
        BOOT_B_CLASS=300, BOOT_B_REG=300,
    )

_cfg = _cfg_by_mode(MODE)

# ===================== LOGGING =====================
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    handlers=[logging.FileHandler(RUN_DIR / "run.log", encoding="utf-8"),
              logging.StreamHandler(sys.stdout)]
)
logger = logging.getLogger(__name__)

def _set_seed(seed=RANDOM_STATE):
    random.seed(seed)
    np.random.seed(seed)

_set_seed()

def save_env_and_config():
    meta = [
        f"Python: {sys.version}",
        f"Platform: {platform.platform()}",
        f"numpy: {np.__version__}",
        f"pandas: {pd.__version__}",
        f"lightgbm: {lgb.__version__}",
        f"xgboost: {(xgb.__version__ if _HAS_XGB else 'NOT_INSTALLED')}",
        f"statsmodels: {('OK' if _HAS_SM else 'NOT_INSTALLED')}",
        f"joblib: {('OK' if _HAS_JOBLIB else 'NOT_INSTALLED')}",
    ]
    ENV_TXT.write_text("\n".join(meta), encoding="utf-8")

    CFG_JSON.write_text(json.dumps(dict(
        MODE=str(MODE),
        cfg=_cfg,
        RANDOM_STATE=int(RANDOM_STATE),
        RAW_PATH=str(RAW_PATH),
        LOS_CUTOFF_DAYS=int(LOS_CUTOFF_DAYS),
        ENABLE_LOSBIN=bool(ENABLE_LOSBIN),
        FEATURES_JSON_HOLDOUT=str(FEATURES_JSON_HOLDOUT),
        FEATURES_JSON_TEMPORAL=str(FEATURES_JSON_TEMPORAL),
        SHAP_DIR=str(SHAP_DIR),
    ), indent=2, ensure_ascii=False), encoding="utf-8")

save_env_and_config()

# ===================== COLUMNS (example set) =====================
FEATURES_BASE = [
    'Ambient temperature','Atmospheric pressure','Age','Male','BMI',
    'NIHSS total',
    'NIHSS: level of consciousness','NIHSS: response to questions','NIHSS: response to commands',
    'NIHSS: best gaze','NIHSS: visual field','NIHSS: facial palsy',
    'NIHSS: motor arm (R)','NIHSS: motor arm (L)',
    'NIHSS: motor leg (R)','NIHSS: motor leg (L)',
    'NIHSS: limb ataxia','NIHSS: sensory','NIHSS: language','NIHSS: dysarthria','NIHSS: extinction/inattention',
    't-PA','Edaravone','Antiepileptic drugs','RAS inhibitors','ARNI',
    'Aspirin','P2Y12 inhibitors','Anticoagulants','Diuretics','SGLT2 inhibitors',
    'Oral antidiabetic drugs','Insulin','Statins','Beta-blockers','Antibiotics',
    'Mechanical ventilation','General anesthesia','Maintenance dialysis',
    'Endovascular thrombectomy','Oral intake',
    'Heart rate','systolic BP','diastolic BP',
    'Other ischemic stroke','Cardioembolic stroke','Intracerebral hemorrhage',
    'Atherothrombotic infarction','Lacunar infarction','Transient ischemic attack',
]
LAB_BASE = ['Albumin','BNP','BUN','CRP','Hemoglobin','Potassium','Sodium','PT-INR','WBC','eGFR']

# ===================== UTIL =====================
def _safe_num(s: pd.Series) -> pd.Series:
    return pd.to_numeric(s, errors="coerce")

def _nanpercentile_safe(a, q):
    a = np.asarray(a, float)
    a = a[np.isfinite(a)]
    if a.size == 0:
        return np.nan
    return float(np.nanpercentile(a, q))

def _log_ratio(a: pd.Series, b: pd.Series) -> pd.Series:
    a = _safe_num(a)
    b = _safe_num(b)
    qa = _nanpercentile_safe(a, 1)
    qb = _nanpercentile_safe(b, 1)
    eps_a = 1e-6 if (not np.isfinite(qa) or qa <= 0) else float(qa) * 0.5
    eps_b = 1e-6 if (not np.isfinite(qb) or qb <= 0) else float(qb) * 0.5
    r = (a + eps_a) / (b + eps_b)
    r = r.where(r > 0)
    with np.errstate(divide="ignore", invalid="ignore"):
        out = np.log(r.astype(float))
    return out.clip(-5, 5)

def build_feature_table(df: pd.DataFrame) -> pd.DataFrame:
    cols = [c for c in FEATURES_BASE if c in df.columns]
    X = df[cols].copy()
    for c in X.columns:
        X[c] = _safe_num(X[c])

    # 48h時点予測：admissionと48hを使うのは整合
    for base in LAB_BASE:
        ad = f"{base} (admission)"
        h2 = f"{base} (48h)"
        if ad in df.columns and h2 in df.columns:
            X[f"log_ratio_{base}"] = _log_ratio(df[h2], df[ad]).astype(float)

    return X.astype(np.float32)

def detect_continuous_cols(X: pd.DataFrame, thr_unique=3) -> List[str]:
    cont = []
    for c in X.columns:
        nunq = int(pd.to_numeric(X[c], errors="coerce").nunique(dropna=True))
        if nunq > thr_unique:
            cont.append(c)
    return cont

def class_weight_vec(y: np.ndarray) -> np.ndarray:
    y = np.asarray(y, int)
    classes = np.unique(y)
    cw = compute_class_weight(class_weight="balanced", classes=classes, y=y)
    m = {c: w for c, w in zip(classes, cw)}
    return np.array([m[v] for v in y], np.float32)

def rmse(a, b) -> float:
    a = np.asarray(a, float)
    b = np.asarray(b, float)
    return float(np.sqrt(np.mean((a - b) ** 2)))

def smape(y_true, y_pred) -> float:
    y_true = np.asarray(y_true, float)
    y_pred = np.asarray(y_pred, float)
    denom = np.abs(y_true) + np.abs(y_pred)
    denom = np.where(denom == 0, 1.0, denom)
    return float(np.mean(2.0 * np.abs(y_pred - y_true) / denom))

def _sigmoid(x: np.ndarray) -> np.ndarray:
    x = np.asarray(x, float)
    x = np.clip(x, -50, 50)
    return 1.0 / (1.0 + np.exp(-x))

def _hash_text(s: str) -> str:
    return hashlib.sha256(s.encode("utf-8")).hexdigest()

def make_cache_signature(feat_cols: List[str]) -> Dict[str, Any]:
    """
    Signature to detect column/cfg changes and prevent silent cache reuse.
    """
    feat_sig = _hash_text("|".join(map(str, feat_cols)))
    cfg_sig = _hash_text(json.dumps(
        dict(MODE=MODE, cfg=_cfg, RANDOM_STATE=RANDOM_STATE, LOS_CUTOFF_DAYS=LOS_CUTOFF_DAYS, ENABLE_LOSBIN=ENABLE_LOSBIN),
        sort_keys=True, ensure_ascii=False
    ))
    raw_sig = dict(
        RAW_PATH=str(RAW_PATH),
        RAW_MTIME=float(os.path.getmtime(RAW_PATH)) if os.path.exists(RAW_PATH) else np.nan,
    )
    return dict(feat_cols_sha256=feat_sig, cfg_sha256=cfg_sig, raw=raw_sig)

# ===================== PREPROCESSOR (winsor/log1p/median/z) =====================
class WinsorLogZ:
    """
    - winsorize at 1st/99th pct
    - log1p for non-negative skewed continuous vars
    - (optional) median imputation
    - z-score for continuous vars
    Fit on training only.

    PATCH: impute=True/False
      - Trees: impute=True (as before)
      - LR+MI: impute=False to preserve missingness for IterativeImputer
    """
    def __init__(self, ql=0.01, qh=0.99, skew_thr=1.0, impute: bool = True):
        self.ql = ql
        self.qh = qh
        self.skew_thr = skew_thr
        self.impute = bool(impute)
        self.cont_cols: List[str] = []
        self.qs: Dict[str, Tuple[float, float]] = {}
        self.logs: set[str] = set()
        self.means: Dict[str, float] = {}
        self.stds: Dict[str, float] = {}
        self.medians: Dict[str, float] = {}

    @staticmethod
    def _clip_q(a: np.ndarray, lo: float, hi: float) -> np.ndarray:
        a = np.asarray(a, float)
        if not (np.isfinite(lo) and np.isfinite(hi)) or hi <= lo:
            m = np.nanmedian(a)
            w = 1e-6 if not np.isfinite(m) else max(1e-6, abs(m) * 1e-3)
            return np.clip(a, m - w, m + w)
        return np.clip(a, lo, hi)

    def fit(self, X: pd.DataFrame):
        X = X.copy()
        self.cont_cols = detect_continuous_cols(X, thr_unique=3)

        for c in X.columns:
            s = pd.to_numeric(X[c], errors="coerce").to_numpy(dtype=np.float32)
            lo = _nanpercentile_safe(s, self.ql * 100)
            hi = _nanpercentile_safe(s, self.qh * 100)
            self.qs[c] = (lo, hi)
            s2 = self._clip_q(s, lo, hi)

            if c in self.cont_cols:
                sk = float(pd.Series(s2).skew(skipna=True)) if np.isfinite(np.nanmean(s2)) else 0.0
                if (np.nanmin(s2) >= 0) and (abs(sk) > self.skew_thr):
                    self.logs.add(c)
                    s2 = np.log1p(s2)

                med = float(np.nanmedian(s2)) if np.isfinite(np.nanmedian(s2)) else 0.0
                self.medians[c] = med

                # stats for z-score must be computed on an imputed vector
                s2_imp = np.where(np.isnan(s2), med, s2)
                m = float(np.nanmean(s2_imp))
                sd = float(np.nanstd(s2_imp))
                if not (np.isfinite(sd) and sd > 0):
                    sd = 1.0
                self.means[c] = m
                self.stds[c] = sd
            else:
                med = float(np.nanmedian(s2)) if np.isfinite(np.nanmedian(s2)) else 0.0
                self.medians[c] = med

        return self

    def transform(self, X: pd.DataFrame) -> pd.DataFrame:
        X = X.copy()
        for c in X.columns:
            s = pd.to_numeric(X[c], errors="coerce").to_numpy(dtype=np.float32)
            lo, hi = self.qs.get(c, (np.nan, np.nan))
            s = self._clip_q(s, lo, hi)
            if c in self.logs:
                s = np.log1p(s)

            med = float(self.medians.get(c, 0.0))
            if self.impute:
                s = np.where(np.isnan(s), med, s)
            # else: keep NaN for MI path

            if c in self.cont_cols:
                # z-score: keep NaN as NaN when impute=False
                m = float(self.means.get(c, 0.0))
                sd = float(self.stds.get(c, 1.0))
                denom = (sd if (np.isfinite(sd) and sd > 0) else 1.0)
                if self.impute:
                    s = (s - m) / denom
                else:
                    # preserve NaN
                    s = np.where(np.isnan(s), np.nan, (s - m) / denom)

            X[c] = s.astype(np.float32)
        return X.astype(np.float32)

# ===================== VIF =====================
def _safe_vif(Xvals: np.ndarray, i: int) -> float:
    if not _HAS_SM:
        y = Xvals[:, i]
        X = np.delete(Xvals, i, axis=1)
        X = np.column_stack([X, np.ones(len(X), dtype=X.dtype)])
        beta, *_ = np.linalg.lstsq(X, y, rcond=None)
        yhat = X @ beta
        sst = float(np.sum((y - y.mean()) ** 2))
        if sst <= 0:
            return np.nan
        ssr = float(np.sum((yhat - y.mean()) ** 2))
        r2 = min(max(ssr / sst, 0.0), 0.999999)
        return 1.0 / (1.0 - r2)
    return float(_sm_vif(Xvals, i))

def vif_filter(
    X: pd.DataFrame,
    thr: float,
    protected: List[str],
    numeric_only: bool = True,
    max_iter: int = 8
) -> List[str]:
    """
    VIF filter on a working copy.
    If numeric_only=True, we apply VIF only to "continuous-ish" cols (nunique>3).
    Protected features are never removed.
    """
    prot = set(protected or [])
    work = X.copy()
    work = work.apply(lambda s: pd.to_numeric(s, errors="coerce")).astype(np.float32)

    if numeric_only:
        cont = set(detect_continuous_cols(work, thr_unique=3))
        keep_vif = [c for c in work.columns if (c in cont) or (c in prot)]
    else:
        keep_vif = list(work.columns)

    work_v = work[keep_vif].copy()
    work_v = work_v.fillna(work_v.median(numeric_only=True))

    if work_v.shape[1] <= 2:
        return list(work.columns)

    for _ in range(max_iter):
        if work_v.shape[1] <= 2:
            break
        cols = list(work_v.columns)
        X_ = work_v.assign(__const__=1.0).astype(np.float32).values
        vifs = []
        for i, c in enumerate(cols):
            try:
                v = _safe_vif(X_, i)
            except Exception:
                v = np.nan
            vifs.append((c, float(v)))
        dfv = pd.DataFrame(vifs, columns=["f", "v"]).sort_values("v", ascending=False)

        cand = dfv[~dfv["f"].isin(prot)]
        if len(cand) and np.isfinite(float(cand.iloc[0]["v"])) and float(cand.iloc[0]["v"]) > thr and work_v.shape[1] >= 3:
            rm = str(cand.iloc[0]["f"])
            logger.info(f"[VIF] removed: {rm}, vif={float(cand.iloc[0]['v']):.2f}")
            work_v = work_v.drop(columns=[rm])
        else:
            break

    keep_set = set(work_v.columns) | (set(work.columns) - set(keep_vif))
    keep_set |= (prot & set(work.columns))
    keep = [c for c in X.columns if c in keep_set]
    return keep

# ===================== CALIBRATION (Platt) =====================
class PlattCalibrator:
    def __init__(self):
        self.lr = None

    def fit(self, y_true: np.ndarray, p: np.ndarray):
        y = np.asarray(y_true, int)
        p = np.clip(np.asarray(p, float), 1e-6, 1 - 1e-6)
        logit = np.log(p / (1 - p)).reshape(-1, 1)
        self.lr = LogisticRegression(solver="lbfgs", max_iter=2000, C=1e6, random_state=RANDOM_STATE)
        self.lr.fit(logit, y)
        return self

    def transform(self, p: np.ndarray) -> np.ndarray:
        if self.lr is None:
            return np.clip(np.asarray(p, float), 1e-6, 1 - 1e-6).astype(np.float32)
        p = np.clip(np.asarray(p, float), 1e-6, 1 - 1e-6)
        logit = np.log(p / (1 - p)).reshape(-1, 1)
        out = self.lr.predict_proba(logit)[:, 1]
        return np.clip(out, 1e-6, 1 - 1e-6).astype(np.float32)

# ===================== METRICS =====================
def citl_intercept_only_fixed_offset(y_true: np.ndarray, p: np.ndarray, max_iter: int = 50, tol: float = 1e-10) -> float:
    """
    CITL with offset fixed to logit(p): estimate intercept b only.
    Solve: sum(y - sigmoid(b + offset)) = 0 using Newton-Raphson.

    This matches the standard "calibration-in-the-large" definition under fixed slope=1.
    """
    y = np.asarray(y_true, float)
    pp = np.clip(np.asarray(p, float), 1e-6, 1 - 1e-6)
    off = np.log(pp / (1 - pp))  # fixed offset

    b = 0.0
    for _ in range(max_iter):
        eta = b + off
        mu = _sigmoid(eta)
        g = float(np.sum(y - mu))                       # gradient
        h = -float(np.sum(mu * (1.0 - mu)))            # Hessian
        if not (np.isfinite(g) and np.isfinite(h)) or h == 0.0:
            break
        step = g / h
        b_new = b - step
        if abs(b_new - b) < tol:
            b = b_new
            break
        b = b_new

    if not np.isfinite(b):
        b = 0.0
    return float(b)

def citl_with_offset(y_true: np.ndarray, p: np.ndarray) -> float:
    pp = np.clip(np.asarray(p, float), 1e-6, 1 - 1e-6)
    yy = np.asarray(y_true, int)
    off = np.log(pp / (1 - pp))

    if _HAS_SM:
        X = np.ones((len(pp), 1))
        try:
            res = sm.GLM(yy, X, family=sm.families.Binomial(), offset=off).fit()
            return float(res.params[0])
        except Exception:
            pass

    # PATCH: sklearn fallback replaced by correct intercept-only Newton solver
    return citl_intercept_only_fixed_offset(yy, pp)

def safe_cls_metrics(y_true: np.ndarray, p1: np.ndarray) -> Dict[str, float]:
    y_true = np.asarray(y_true, int)
    p1 = np.asarray(p1, float)
    out = {
        "AUROC": np.nan,
        "AP": np.nan,
        "Brier": float(np.mean((p1 - y_true) ** 2)),
        "CalibSlope": np.nan,
        "CITL": np.nan,
    }
    if np.unique(y_true).size < 2:
        return out

    out["AUROC"] = float(roc_auc_score(y_true, p1))
    out["AP"] = float(average_precision_score(y_true, p1))

    pp = np.clip(p1, 1e-6, 1 - 1e-6)
    logit = np.log(pp / (1 - pp)).reshape(-1, 1)
    if np.isfinite(logit).all() and np.var(logit) > 0:
        lr = LogisticRegression(C=1e6, solver="lbfgs", max_iter=2000).fit(logit, y_true)
        out["CalibSlope"] = float(lr.coef_.ravel()[0])
        out["CITL"] = float(citl_with_offset(y_true, p1))
    return out

# ===================== SMearing (log1p regression) =====================
def smearing_factor(y_true: np.ndarray, yhat_log: np.ndarray) -> float:
    yt = np.asarray(y_true, float)
    ylog = np.log1p(np.clip(yt, 0, None))
    resid = ylog - np.asarray(yhat_log, float)
    sf = float(np.mean(np.exp(resid[np.isfinite(resid)]))) if np.any(np.isfinite(resid)) else 1.0
    if not np.isfinite(sf) or sf <= 0:
        sf = 1.0
    return sf

# ===================== FEATURE RANKING (XGB preferred; fallback LGBM) =====================
def gain_rank_features(
    X_tr: pd.DataFrame,
    y_tr: np.ndarray,
    task: str,  # "binary" | "reg"
    random_state: int,
) -> pd.Series:
    """
    Returns gain-based importance ranking.
    - Prefer XGB gain (if available)
    - Else fallback to LGBM importance
    """
    X = X_tr.copy()
    X = X.apply(lambda s: pd.to_numeric(s, errors="coerce")).astype(np.float32)
    X = X.fillna(X.median(numeric_only=True))

    if _HAS_XGB:
        if task == "binary":
            model = xgb.XGBClassifier(
                n_estimators=500,
                max_depth=4,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                reg_alpha=0.0,
                reg_lambda=1.0,
                objective="binary:logistic",
                n_jobs=1,
                random_state=random_state,
                eval_metric="logloss",
            )
            sw = class_weight_vec(y_tr)
            model.fit(X.to_numpy(), y_tr, sample_weight=sw)
        else:
            model = xgb.XGBRegressor(
                n_estimators=500,
                max_depth=4,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                reg_alpha=0.0,
                reg_lambda=1.0,
                objective="reg:squarederror",
                n_jobs=1,
                random_state=random_state,
            )
            model.fit(X.to_numpy(), np.log1p(np.asarray(y_tr, float)))

        booster = model.get_booster()
        score = booster.get_score(importance_type="gain")

        imp = np.zeros(X.shape[1], float)
        for k, v in score.items():
            if k.startswith("f"):
                idx = int(k[1:])
                if 0 <= idx < len(imp):
                    imp[idx] = float(v)
        return pd.Series(imp, index=list(X.columns)).sort_values(ascending=False)

    # fallback: LGBM
    if task == "binary":
        clf = lgb.LGBMClassifier(
            objective="binary",
            n_estimators=800,
            learning_rate=0.05,
            num_leaves=63,
            subsample=0.8,
            colsample_bytree=0.8,
            n_jobs=1,
            random_state=random_state,
        )
        sw = class_weight_vec(y_tr)
        clf.fit(X.to_numpy(), y_tr, sample_weight=sw)
        imp = clf.feature_importances_.astype(float)
        return pd.Series(imp, index=list(X.columns)).sort_values(ascending=False)
    else:
        reg = lgb.LGBMRegressor(
            objective="regression",
            n_estimators=800,
            learning_rate=0.05,
            num_leaves=63,
            subsample=0.8,
            colsample_bytree=0.8,
            n_jobs=1,
            random_state=random_state,
        )
        reg.fit(X.to_numpy(), np.log1p(np.asarray(y_tr, float)))
        imp = reg.feature_importances_.astype(float)
        return pd.Series(imp, index=list(X.columns)).sort_values(ascending=False)

def select_features_once_trainonly(
    X_train: pd.DataFrame,
    y_train: np.ndarray,
    task: str,  # "binary" | "reg"
    topk: int,
    protected: List[str],
    vif_thr: float,
    vif_numeric_only: bool = True,
) -> List[str]:
    """
    Train-only, once-per-task feature selection:
    - gain rank -> topk
    - force protected
    - VIF filter
    """
    cols = list(X_train.columns)
    if len(cols) <= max(2, topk):
        keep = cols.copy()
    else:
        s = gain_rank_features(X_train, y_train, task=task, random_state=RANDOM_STATE + 901)
        keep = list(s.index[: min(topk, len(s))])

    for p in (protected or []):
        if p in cols and p not in keep:
            keep = [p] + keep

    keep = [c for c in cols if c in set(keep)]  # stable order
    keep = vif_filter(X_train[keep], thr=vif_thr, protected=protected,
                      numeric_only=vif_numeric_only, max_iter=8)
    keep = [c for c in cols if c in set(keep)]  # restore original order
    return keep

# ===================== MODEL FACTORIES =====================
def make_lgbm_cls_base() -> lgb.LGBMClassifier:
    return lgb.LGBMClassifier(
        objective="binary",
        n_estimators=_cfg["LGBM_N_EST_CLS"],
        learning_rate=0.05,
        num_leaves=63,
        max_depth=-1,
        min_child_samples=20,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.0,
        reg_lambda=0.0,
        n_jobs=1,
        random_state=RANDOM_STATE,
    )

def make_lgbm_reg_base(objective="regression") -> lgb.LGBMRegressor:
    return lgb.LGBMRegressor(
        objective=objective,
        n_estimators=_cfg["LGBM_N_EST_REG"],
        learning_rate=0.05,
        num_leaves=63,
        max_depth=-1,
        min_child_samples=20,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.0,
        reg_lambda=0.0,
        n_jobs=1,
        random_state=RANDOM_STATE,
    )

def make_lr_base() -> LogisticRegression:
    return LogisticRegression(
        penalty="elasticnet",
        solver="saga",
        l1_ratio=0.5,
        C=1.0,
        max_iter=4000,
        n_jobs=1,
        random_state=RANDOM_STATE,
    )

def _lgb_callbacks():
    cb = []
    try:
        cb.append(lgb.log_evaluation(period=0))
    except Exception:
        pass
    try:
        es_rounds = int(max(10, _cfg.get("EARLY_STOP", 50)))
        cb.append(lgb.early_stopping(stopping_rounds=es_rounds, verbose=False))
    except Exception:
        pass
    return cb

def fit_lgbm_classifier(model: lgb.LGBMClassifier, Xtr: np.ndarray, ytr: np.ndarray, sw: Optional[np.ndarray]):
    tr_i, va_i = train_test_split(
        np.arange(len(ytr)),
        test_size=0.15,
        stratify=ytr,
        random_state=RANDOM_STATE + 123,
    )
    X_tr, X_va = Xtr[tr_i], Xtr[va_i]
    y_tr, y_va = ytr[tr_i], ytr[va_i]
    kwargs = dict(eval_set=[(X_va, y_va)], eval_metric="binary_logloss", callbacks=_lgb_callbacks())
    if sw is not None:
        model.fit(X_tr, y_tr, sample_weight=sw[tr_i], **kwargs)
    else:
        model.fit(X_tr, y_tr, **kwargs)
    return model

def fit_lgbm_regressor(model: lgb.LGBMRegressor, Xtr: np.ndarray, ytr_log: np.ndarray):
    tr_i, va_i = train_test_split(
        np.arange(len(ytr_log)),
        test_size=0.15,
        random_state=RANDOM_STATE + 321,
    )
    X_tr, X_va = Xtr[tr_i], Xtr[va_i]
    y_tr, y_va = ytr_log[tr_i], ytr_log[va_i]
    model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], eval_metric="rmse", callbacks=_lgb_callbacks())
    return model

# ===================== OPTUNA CORE =====================
def _make_study(direction="maximize", name="study"):
    return optuna.create_study(
        direction=direction,
        study_name=f"{name}_{MODE}_seed{RANDOM_STATE}",
        sampler=TPESampler(seed=RANDOM_STATE, multivariate=False),
        pruner=MedianPruner(n_startup_trials=8, n_warmup_steps=0, interval_steps=1),
    )

def subsample_indices(n: int, frac: float, y: Optional[np.ndarray] = None) -> np.ndarray:
    m = n if frac >= 1.0 else max(800, int(n * frac))
    rng = np.random.default_rng(RANDOM_STATE)
    if y is None or np.unique(y).size < 2:
        return np.sort(rng.choice(n, size=min(m, n), replace=False))
    idx_all = np.arange(n)
    uniq, counts = np.unique(y, return_counts=True)
    take = {uniq[i]: max(2, int(m * (counts[i] / n))) for i in range(len(uniq))}
    chosen = []
    for c in uniq:
        pool = idx_all[y == c]
        k = min(take[c], len(pool))
        if k > 0:
            chosen.append(rng.choice(pool, size=k, replace=False))
    return np.sort(np.concatenate(chosen))

def _clip_nearest(n_features: int, n_nearest: int) -> int:
    # IterativeImputer requirement: 1 <= n_nearest_features <= n_features
    if n_features <= 1:
        return 1
    return int(max(1, min(n_nearest, n_features - 1)))

# ===================== TUNING (fixed FS for the task) =====================
def tune_xgb_cls_fixedfs(X: pd.DataFrame, y: np.ndarray, keep: List[str], metric: str = "AUROC") -> dict:
    assert metric in ("AUROC", "AP")
    if not _HAS_XGB:
        raise RuntimeError("xgboost not installed (tune_xgb).")
    sub_idx = subsample_indices(len(y), _cfg["TUNE_SUBSAMPLE_FRAC"], y)
    Xsub, ysub = X.iloc[sub_idx][keep].reset_index(drop=True), y[sub_idx]
    cv = StratifiedKFold(n_splits=_cfg["N_SPLITS_CLS"], shuffle=True, random_state=RANDOM_STATE)

    def objective(trial: optuna.Trial) -> float:
        params = {
            "max_depth": trial.suggest_int("max_depth", 3, 8),
            "learning_rate": trial.suggest_float("learning_rate", 0.02, 0.2, log=True),
            "subsample": trial.suggest_float("subsample", 0.6, 0.95),
            "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 0.95),
            "reg_alpha": trial.suggest_float("reg_alpha", 1e-6, 1.0, log=True),
            "reg_lambda": trial.suggest_float("reg_lambda", 1e-6, 5.0, log=True),
            "min_child_weight": trial.suggest_float("min_child_weight", 1e-3, 10.0, log=True),
        }
        oof = np.zeros(len(ysub), np.float32)

        for fold, (tr, va) in enumerate(cv.split(Xsub, ysub), start=1):
            Xtr_raw, Xva_raw = Xsub.iloc[tr].copy(), Xsub.iloc[va].copy()
            ytr, yva = ysub[tr], ysub[va]

            pre = WinsorLogZ(impute=True).fit(Xtr_raw)
            Xtr_t = pre.transform(Xtr_raw).to_numpy(dtype=np.float32)
            Xva_t = pre.transform(Xva_raw).to_numpy(dtype=np.float32)

            sw = class_weight_vec(ytr)
            clf = xgb.XGBClassifier(
                n_estimators=_cfg["XGB_N_EST_CLS"],
                objective="binary:logistic",
                n_jobs=1,
                random_state=RANDOM_STATE + fold,
                eval_metric="logloss",
                **params,
            )
            clf.fit(Xtr_t, ytr, sample_weight=sw)
            oof[va] = clf.predict_proba(Xva_t)[:, 1].astype(np.float32)

        if np.unique(ysub).size < 2:
            return np.nan
        return float(roc_auc_score(ysub, oof)) if metric == "AUROC" else float(average_precision_score(ysub, oof))

    study = _make_study(direction="maximize", name=f"tune_XGB_cls_{metric}")
    study.optimize(objective, timeout=_cfg["TUNE_TIMEOUT_SEC"], n_trials=min(_cfg["N_TRIALS"], 80),
                   n_jobs=1, gc_after_trial=True)
    return study.best_params

def tune_lgbm_cls_fixedfs(X: pd.DataFrame, y: np.ndarray, keep: List[str], metric: str = "AUROC") -> dict:
    assert metric in ("AUROC", "AP")
    sub_idx = subsample_indices(len(y), _cfg["TUNE_SUBSAMPLE_FRAC"], y)
    Xsub, ysub = X.iloc[sub_idx][keep].reset_index(drop=True), y[sub_idx]
    cv = StratifiedKFold(n_splits=_cfg["N_SPLITS_CLS"], shuffle=True, random_state=RANDOM_STATE)

    def objective(trial: optuna.Trial) -> float:
        params = {
            "learning_rate": trial.suggest_float("learning_rate", 0.02, 0.10, log=True),
            "num_leaves": trial.suggest_int("num_leaves", 31, 127),
            "min_child_samples": trial.suggest_int("min_child_samples", 10, 60),
            "subsample": trial.suggest_float("subsample", 0.6, 0.95),
            "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 0.95),
            "reg_alpha": trial.suggest_float("reg_alpha", 1e-6, 1.0, log=True),
            "reg_lambda": trial.suggest_float("reg_lambda", 1e-6, 2.0, log=True),
            "max_depth": trial.suggest_int("max_depth", -1, 14),
        }
        oof = np.zeros(len(ysub), np.float32)

        for fold, (tr, va) in enumerate(cv.split(Xsub, ysub), start=1):
            Xtr_raw, Xva_raw = Xsub.iloc[tr].copy(), Xsub.iloc[va].copy()
            ytr, yva = ysub[tr], ysub[va]

            pre = WinsorLogZ(impute=True).fit(Xtr_raw)
            Xtr_t = pre.transform(Xtr_raw).to_numpy(dtype=np.float32)
            Xva_t = pre.transform(Xva_raw).to_numpy(dtype=np.float32)

            clf = make_lgbm_cls_base()
            clf.set_params(**{k: v for k, v in params.items() if k in clf.get_params()})
            clf.set_params(n_jobs=1, random_state=RANDOM_STATE + fold)

            sw = class_weight_vec(ytr)
            clf = fit_lgbm_classifier(clf, Xtr_t, ytr, sw)
            oof[va] = clf.predict_proba(Xva_t)[:, 1].astype(np.float32)

        if np.unique(ysub).size < 2:
            return np.nan
        return float(roc_auc_score(ysub, oof)) if metric == "AUROC" else float(average_precision_score(ysub, oof))

    study = _make_study(direction="maximize", name=f"tune_LGBM_cls_{metric}")
    study.optimize(objective, timeout=_cfg["TUNE_TIMEOUT_SEC"], n_trials=_cfg["N_TRIALS"],
                   n_jobs=1, gc_after_trial=True)
    return study.best_params

def tune_xgb_reg_fixedfs(X: pd.DataFrame, y: np.ndarray, keep: List[str]) -> dict:
    if not _HAS_XGB:
        raise RuntimeError("xgboost not installed (tune_xgb_reg).")
    sub_idx = subsample_indices(len(y), _cfg["TUNE_SUBSAMPLE_FRAC"], None)
    Xsub, ysub = X.iloc[sub_idx][keep].reset_index(drop=True), y[sub_idx]
    cv = KFold(n_splits=_cfg["N_SPLITS_REG"], shuffle=True, random_state=RANDOM_STATE)

    def objective(trial: optuna.Trial) -> float:
        params = {
            "max_depth": trial.suggest_int("max_depth", 3, 8),
            "learning_rate": trial.suggest_float("learning_rate", 0.02, 0.2, log=True),
            "subsample": trial.suggest_float("subsample", 0.6, 0.95),
            "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 0.95),
            "reg_alpha": trial.suggest_float("reg_alpha", 1e-6, 1.0, log=True),
            "reg_lambda": trial.suggest_float("reg_lambda", 1e-6, 5.0, log=True),
            "min_child_weight": trial.suggest_float("min_child_weight", 1e-3, 10.0, log=True),
        }
        oof = np.zeros(len(ysub), np.float32)

        for fold, (tr, va) in enumerate(cv.split(Xsub), start=1):
            Xtr_raw, Xva_raw = Xsub.iloc[tr].copy(), Xsub.iloc[va].copy()
            ytr, yva = ysub[tr], ysub[va]

            pre = WinsorLogZ(impute=True).fit(Xtr_raw)
            Xtr_t = pre.transform(Xtr_raw).to_numpy(dtype=np.float32)
            Xva_t = pre.transform(Xva_raw).to_numpy(dtype=np.float32)

            ytr_log = np.log1p(np.clip(ytr, 0, None))
            reg = xgb.XGBRegressor(
                n_estimators=_cfg["XGB_N_EST_REG"],
                objective="reg:squarederror",
                n_jobs=1,
                random_state=RANDOM_STATE + fold,
                **params,
            )
            reg.fit(Xtr_t, ytr_log)
            pred_log_va = reg.predict(Xva_t).astype(np.float32)
            pred_log_tr = reg.predict(Xtr_t).astype(np.float32)
            sf = smearing_factor(ytr, pred_log_tr)
            oof[va] = (np.exp(pred_log_va) * sf - 1.0).clip(0, None).astype(np.float32)

        return -rmse(ysub, oof)

    study = _make_study(direction="maximize", name="tune_XGB_reg")
    study.optimize(objective, timeout=_cfg["TUNE_TIMEOUT_SEC"], n_trials=min(_cfg["N_TRIALS"], 80),
                   n_jobs=1, gc_after_trial=True)
    return study.best_params

def tune_lgbm_reg_fixedfs(X: pd.DataFrame, y: np.ndarray, keep: List[str], objective_name="regression") -> dict:
    sub_idx = subsample_indices(len(y), _cfg["TUNE_SUBSAMPLE_FRAC"], None)
    Xsub, ysub = X.iloc[sub_idx][keep].reset_index(drop=True), y[sub_idx]
    cv = KFold(n_splits=_cfg["N_SPLITS_REG"], shuffle=True, random_state=RANDOM_STATE)

    def objective(trial: optuna.Trial) -> float:
        params = {
            "learning_rate": trial.suggest_float("learning_rate", 0.02, 0.10, log=True),
            "num_leaves": trial.suggest_int("num_leaves", 31, 127),
            "min_child_samples": trial.suggest_int("min_child_samples", 10, 60),
            "subsample": trial.suggest_float("subsample", 0.6, 0.95),
            "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 0.95),
            "reg_alpha": trial.suggest_float("reg_alpha", 1e-6, 1.0, log=True),
            "reg_lambda": trial.suggest_float("reg_lambda", 1e-6, 2.0, log=True),
            "max_depth": trial.suggest_int("max_depth", -1, 14),
        }
        oof = np.zeros(len(ysub), np.float32)

        for fold, (tr, va) in enumerate(cv.split(Xsub), start=1):
            Xtr_raw, Xva_raw = Xsub.iloc[tr].copy(), Xsub.iloc[va].copy()
            ytr, yva = ysub[tr], ysub[va]

            pre = WinsorLogZ(impute=True).fit(Xtr_raw)
            Xtr_t = pre.transform(Xtr_raw).to_numpy(dtype=np.float32)
            Xva_t = pre.transform(Xva_raw).to_numpy(dtype=np.float32)

            ytr_log = np.log1p(np.clip(ytr, 0, None))
            reg = make_lgbm_reg_base(objective=objective_name)
            reg.set_params(**{k: v for k, v in params.items() if k in reg.get_params()})
            reg.set_params(n_jobs=1, random_state=RANDOM_STATE + fold)
            reg = fit_lgbm_regressor(reg, Xtr_t, ytr_log)

            pred_log_va = reg.predict(Xva_t).astype(np.float32)
            pred_log_tr = reg.predict(Xtr_t).astype(np.float32)
            sf = smearing_factor(ytr, pred_log_tr)
            oof[va] = (np.exp(pred_log_va) * sf - 1.0).clip(0, None).astype(np.float32)

        return -rmse(ysub, oof)

    study = _make_study(direction="maximize", name=f"tune_LGBM_reg_{objective_name}")
    study.optimize(objective, timeout=_cfg["TUNE_TIMEOUT_SEC"], n_trials=_cfg["N_TRIALS"],
                   n_jobs=1, gc_after_trial=True)
    return study.best_params

def tune_lr_with_mi_fixedfs(X: pd.DataFrame, y: np.ndarray, keep: List[str], M: int) -> dict:
    """
    LR tuning with leak-safe MI inside CV (fixed FS).
    Optimize AUROC.

    PATCH: WinsorLogZ(impute=False) to preserve missingness for IterativeImputer.
    """
    sub_idx = subsample_indices(len(y), _cfg["TUNE_SUBSAMPLE_FRAC"], y)
    Xsub, ysub = X.iloc[sub_idx][keep].reset_index(drop=True), y[sub_idx]
    cv = StratifiedKFold(n_splits=_cfg["N_SPLITS_CLS"], shuffle=True, random_state=RANDOM_STATE)

    def objective(trial: optuna.Trial) -> float:
        C = trial.suggest_float("C", 1e-3, 50.0, log=True)
        l1_ratio = trial.suggest_float("l1_ratio", 0.0, 1.0)

        oof = np.zeros(len(ysub), np.float32)

        for fold, (tr, va) in enumerate(cv.split(Xsub, ysub), start=1):
            Xtr_raw, Xva_raw = Xsub.iloc[tr].copy(), Xsub.iloc[va].copy()
            ytr, yva = ysub[tr], ysub[va]

            pre = WinsorLogZ(impute=False).fit(Xtr_raw)
            Xtr_p = pre.transform(Xtr_raw)
            Xva_p = pre.transform(Xva_raw)

            preds = []
            nfeat = Xtr_p.shape[1]
            nn = _clip_nearest(nfeat, int(_cfg["MI_NEAREST"]))

            for m in range(1, M + 1):
                imp = IterativeImputer(
                    random_state=RANDOM_STATE + 10000 + fold * 100 + m,
                    max_iter=_cfg["MI_MAX_ITER"],
                    sample_posterior=True,
                    skip_complete=False,  # ensure MI executes
                    n_nearest_features=nn,
                    initial_strategy="median",
                )
                imp.fit(Xtr_p.to_numpy(dtype=np.float32))
                Xtr_i = imp.transform(Xtr_p.to_numpy(dtype=np.float32)).astype(np.float32)
                Xva_i = imp.transform(Xva_p.to_numpy(dtype=np.float32)).astype(np.float32)

                lr = make_lr_base()
                lr.set_params(C=C, l1_ratio=l1_ratio)
                sw = class_weight_vec(ytr)
                lr.fit(Xtr_i, ytr, sample_weight=sw)
                preds.append(lr.predict_proba(Xva_i)[:, 1].astype(np.float32))

            oof[va] = np.mean(np.stack(preds, axis=0), axis=0).astype(np.float32)

        return float(roc_auc_score(ysub, oof)) if np.unique(ysub).size >= 2 else np.nan

    study = _make_study(direction="maximize", name="tune_LR_MI")
    n_trials = min(_cfg["N_TRIALS"], 80) if MODE.upper() == "MAX" else min(_cfg["N_TRIALS"], 30)
    study.optimize(objective, timeout=_cfg["TUNE_TIMEOUT_SEC"], n_trials=n_trials,
                   n_jobs=1, gc_after_trial=True)
    return study.best_params

# ===================== OOF + TEST (fixed FS) =====================
def oof_and_test_tree_classifier_fixedfs(
    Xtr: pd.DataFrame, ytr: np.ndarray,
    Xte: pd.DataFrame, yte: np.ndarray,
    keep: List[str],
    model_name: str,  # "XGB" | "LGBM"
    best_params: dict,
) -> Tuple[np.ndarray, np.ndarray, PlattCalibrator, Dict[str, Any]]:
    cv = StratifiedKFold(n_splits=_cfg["N_SPLITS_CLS"], shuffle=True, random_state=RANDOM_STATE)
    oof = np.zeros(len(Xtr), np.float32)

    for fold, (tr, va) in enumerate(cv.split(Xtr, ytr), start=1):
        Xtr_raw, Xva_raw = Xtr.iloc[tr][keep].copy(), Xtr.iloc[va][keep].copy()
        y_tr, y_va = ytr[tr], ytr[va]

        pre = WinsorLogZ(impute=True).fit(Xtr_raw)
        Xtr_t = pre.transform(Xtr_raw).to_numpy(dtype=np.float32)
        Xva_t = pre.transform(Xva_raw).to_numpy(dtype=np.float32)

        if model_name == "XGB":
            if not _HAS_XGB:
                raise RuntimeError("xgboost not installed.")
            clf = xgb.XGBClassifier(
                n_estimators=_cfg["XGB_N_EST_CLS"],
                objective="binary:logistic",
                n_jobs=1,
                random_state=RANDOM_STATE + fold,
                eval_metric="logloss",
                **best_params,
            )
            sw = class_weight_vec(y_tr)
            clf.fit(Xtr_t, y_tr, sample_weight=sw)
            oof[va] = clf.predict_proba(Xva_t)[:, 1].astype(np.float32)

        elif model_name == "LGBM":
            clf = make_lgbm_cls_base()
            clf.set_params(**{k: v for k, v in best_params.items() if k in clf.get_params()})
            clf.set_params(n_jobs=1, random_state=RANDOM_STATE + fold)
            sw = class_weight_vec(y_tr)
            clf = fit_lgbm_classifier(clf, Xtr_t, y_tr, sw)
            oof[va] = clf.predict_proba(Xva_t)[:, 1].astype(np.float32)
        else:
            raise ValueError("model_name must be XGB or LGBM")

    calibrator = PlattCalibrator().fit(ytr, oof)
    oof_cal = calibrator.transform(oof)

    Xtr_sel = Xtr[keep].copy()
    Xte_sel = Xte[keep].copy()
    pre2 = WinsorLogZ(impute=True).fit(Xtr_sel)
    Xtr_t2 = pre2.transform(Xtr_sel).to_numpy(dtype=np.float32)
    Xte_t2 = pre2.transform(Xte_sel).to_numpy(dtype=np.float32)

    if model_name == "XGB":
        final = xgb.XGBClassifier(
            n_estimators=_cfg["XGB_N_EST_CLS"],
            objective="binary:logistic",
            n_jobs=1,
            random_state=RANDOM_STATE + 999,
            eval_metric="logloss",
            **best_params,
        )
        sw_full = class_weight_vec(ytr)
        final.fit(Xtr_t2, ytr, sample_weight=sw_full)
        p_te = final.predict_proba(Xte_t2)[:, 1].astype(np.float32)
    else:
        final = make_lgbm_cls_base()
        final.set_params(**{k: v for k, v in best_params.items() if k in final.get_params()})
        final.set_params(n_jobs=1, random_state=RANDOM_STATE + 999)
        sw_full = class_weight_vec(ytr)
        final = fit_lgbm_classifier(final, Xtr_t2, ytr, sw_full)
        p_te = final.predict_proba(Xte_t2)[:, 1].astype(np.float32)

    te_cal = calibrator.transform(p_te)

    extras = {
        "final_model": final,
        "final_preprocessor": pre2,
        "final_features": keep,
        "calibrator": calibrator,
        "oof_raw": oof.astype(np.float32),
    }
    return oof_cal, te_cal, calibrator, extras

def oof_and_test_tree_regressor_fixedfs(
    Xtr: pd.DataFrame, ytr: np.ndarray,
    Xte: pd.DataFrame, yte: np.ndarray,
    keep: List[str],
    model_name: str,  # "XGB" | "LGBM"
    best_params: dict,
    reg_objective: str = "regression",
) -> Tuple[np.ndarray, np.ndarray, Dict[str, Any]]:
    cv = KFold(n_splits=_cfg["N_SPLITS_REG"], shuffle=True, random_state=RANDOM_STATE)
    oof = np.zeros(len(Xtr), np.float32)

    for fold, (tr, va) in enumerate(cv.split(Xtr), start=1):
        Xtr_raw, Xva_raw = Xtr.iloc[tr][keep].copy(), Xtr.iloc[va][keep].copy()
        y_tr, y_va = ytr[tr], ytr[va]

        pre = WinsorLogZ(impute=True).fit(Xtr_raw)
        Xtr_t = pre.transform(Xtr_raw).to_numpy(dtype=np.float32)
        Xva_t = pre.transform(Xva_raw).to_numpy(dtype=np.float32)

        ytr_log = np.log1p(np.clip(y_tr, 0, None))

        if model_name == "XGB":
            if not _HAS_XGB:
                raise RuntimeError("xgboost not installed.")
            reg = xgb.XGBRegressor(
                n_estimators=_cfg["XGB_N_EST_REG"],
                objective="reg:squarederror",
                n_jobs=1,
                random_state=RANDOM_STATE + fold,
                **best_params,
            )
            reg.fit(Xtr_t, ytr_log)
            pred_log_va = reg.predict(Xva_t).astype(np.float32)
            pred_log_tr = reg.predict(Xtr_t).astype(np.float32)
        elif model_name == "LGBM":
            reg = make_lgbm_reg_base(objective=reg_objective)
            reg.set_params(**{k: v for k, v in best_params.items() if k in reg.get_params()})
            reg.set_params(n_jobs=1, random_state=RANDOM_STATE + fold)
            reg = fit_lgbm_regressor(reg, Xtr_t, ytr_log)
            pred_log_va = reg.predict(Xva_t).astype(np.float32)
            pred_log_tr = reg.predict(Xtr_t).astype(np.float32)
        else:
            raise ValueError("model_name must be XGB or LGBM")

        sf = smearing_factor(y_tr, pred_log_tr)
        oof[va] = (np.exp(pred_log_va) * sf - 1.0).clip(0, None).astype(np.float32)

    Xtr_sel = Xtr[keep].copy()
    Xte_sel = Xte[keep].copy()
    pre2 = WinsorLogZ(impute=True).fit(Xtr_sel)
    Xtr_t2 = pre2.transform(Xtr_sel).to_numpy(dtype=np.float32)
    Xte_t2 = pre2.transform(Xte_sel).to_numpy(dtype=np.float32)

    ytr_log2 = np.log1p(np.clip(ytr, 0, None))

    if model_name == "XGB":
        final = xgb.XGBRegressor(
            n_estimators=_cfg["XGB_N_EST_REG"],
            objective="reg:squarederror",
            n_jobs=1,
            random_state=RANDOM_STATE + 999,
            **best_params,
        )
        final.fit(Xtr_t2, ytr_log2)
        pred_log_te = final.predict(Xte_t2).astype(np.float32)
        pred_log_tr = final.predict(Xtr_t2).astype(np.float32)
    else:
        final = make_lgbm_reg_base(objective=reg_objective)
        final.set_params(**{k: v for k, v in best_params.items() if k in final.get_params()})
        final.set_params(n_jobs=1, random_state=RANDOM_STATE + 999)
        final = fit_lgbm_regressor(final, Xtr_t2, ytr_log2)
        pred_log_te = final.predict(Xte_t2).astype(np.float32)
        pred_log_tr = final.predict(Xtr_t2).astype(np.float32)

    sf_full = smearing_factor(ytr, pred_log_tr)
    pred_te = (np.exp(pred_log_te) * sf_full - 1.0).clip(0, None).astype(np.float32)

    extras = {
        "final_model": final,
        "final_preprocessor": pre2,
        "final_features": keep,
        "smear_full": float(sf_full),
        "oof_pred": oof.astype(np.float32),
    }
    return oof.astype(np.float32), pred_te.astype(np.float32), extras

def oof_and_test_lr_with_mi_fixedfs(
    Xtr: pd.DataFrame, ytr: np.ndarray,
    Xte: pd.DataFrame, yte: np.ndarray,
    keep: List[str],
    best_params: dict,
    M: int,
) -> Tuple[List[np.ndarray], List[np.ndarray], List[Dict[str, Any]]]:
    """
    LR + MI with fixed FS.
    For each imputation m:
      - OOF via CV: preproc fit fold-train -> imputer fit fold-train -> LR -> fold-val probs
      - Platt fit on OOF
      - Final: preproc fit full train -> imputer fit full train -> LR -> test probs -> Platt

    PATCH: WinsorLogZ(impute=False) so MI is effective.
    """
    cv = StratifiedKFold(n_splits=_cfg["N_SPLITS_CLS"], shuffle=True, random_state=RANDOM_STATE)
    Xtr_use = Xtr[keep].copy()
    Xte_use = Xte[keep].copy()

    oof_list, te_list, extras_list = [], [], []

    for m in range(1, M + 1):
        oof = np.zeros(len(Xtr_use), np.float32)

        for fold, (tr, va) in enumerate(cv.split(Xtr_use, ytr), start=1):
            Xtr_raw, Xva_raw = Xtr_use.iloc[tr].copy(), Xtr_use.iloc[va].copy()
            y_tr, y_va = ytr[tr], ytr[va]

            pre = WinsorLogZ(impute=False).fit(Xtr_raw)
            Xtr_p = pre.transform(Xtr_raw)
            Xva_p = pre.transform(Xva_raw)

            nfeat = Xtr_p.shape[1]
            nn = _clip_nearest(nfeat, int(_cfg["MI_NEAREST"]))

            imp = IterativeImputer(
                random_state=RANDOM_STATE + 20000 + fold * 100 + m,
                max_iter=_cfg["MI_MAX_ITER"],
                sample_posterior=True,
                skip_complete=False,
                n_nearest_features=nn,
                initial_strategy="median",
            )
            imp.fit(Xtr_p.to_numpy(dtype=np.float32))
            Xtr_i = imp.transform(Xtr_p.to_numpy(dtype=np.float32)).astype(np.float32)
            Xva_i = imp.transform(Xva_p.to_numpy(dtype=np.float32)).astype(np.float32)

            lr = make_lr_base()
            lr.set_params(C=float(best_params.get("C", 1.0)), l1_ratio=float(best_params.get("l1_ratio", 0.5)))
            sw = class_weight_vec(y_tr)
            lr.fit(Xtr_i, y_tr, sample_weight=sw)
            oof[va] = lr.predict_proba(Xva_i)[:, 1].astype(np.float32)

        cal = PlattCalibrator().fit(ytr, oof)
        oof_cal = cal.transform(oof)

        pre2 = WinsorLogZ(impute=False).fit(Xtr_use)
        Xtr_p2 = pre2.transform(Xtr_use)
        Xte_p2 = pre2.transform(Xte_use)

        nfeat2 = Xtr_p2.shape[1]
        nn2 = _clip_nearest(nfeat2, int(_cfg["MI_NEAREST"]))

        imp2 = IterativeImputer(
            random_state=RANDOM_STATE + 30000 + m,
            max_iter=_cfg["MI_MAX_ITER"],
            sample_posterior=True,
            skip_complete=False,
            n_nearest_features=nn2,
            initial_strategy="median",
        )
        imp2.fit(Xtr_p2.to_numpy(dtype=np.float32))
        Xtr_i2 = imp2.transform(Xtr_p2.to_numpy(dtype=np.float32)).astype(np.float32)
        Xte_i2 = imp2.transform(Xte_p2.to_numpy(dtype=np.float32)).astype(np.float32)

        lr2 = make_lr_base()
        lr2.set_params(C=float(best_params.get("C", 1.0)), l1_ratio=float(best_params.get("l1_ratio", 0.5)))
        sw_full = class_weight_vec(ytr)
        lr2.fit(Xtr_i2, ytr, sample_weight=sw_full)

        p_te = lr2.predict_proba(Xte_i2)[:, 1].astype(np.float32)
        p_te_cal = cal.transform(p_te)

        oof_list.append(oof_cal.astype(np.float32))
        te_list.append(p_te_cal.astype(np.float32))
        extras_list.append({
            "final_model": lr2,
            "final_preprocessor": pre2,
            "final_imputer": imp2,
            "final_features": keep,
            "calibrator": cal,
            "oof_raw": oof.astype(np.float32),
        })

        gc.collect()

    return oof_list, te_list, extras_list

# ===================== BOOTSTRAP + RUBIN =====================
def paired_boot_var_cls(y, p, metric="AUROC", B=1000, seed=0) -> float:
    rng = np.random.default_rng(seed)
    y = np.asarray(y, int)
    p = np.asarray(p, float)
    n = len(y)
    vals = []
    for _ in range(B):
        idx = rng.integers(0, n, size=n)
        yy = y[idx]
        pp = p[idx]
        if metric == "AUROC":
            if np.unique(yy).size < 2:
                continue
            vals.append(roc_auc_score(yy, pp))
        elif metric == "AP":
            if np.unique(yy).size < 2:
                continue
            vals.append(average_precision_score(yy, pp))
        elif metric == "Brier":
            vals.append(float(np.mean((pp - yy) ** 2)))
        elif metric == "CalibSlope":
            vals.append(float(safe_cls_metrics(yy, pp)["CalibSlope"]))
        elif metric == "CITL":
            vals.append(float(safe_cls_metrics(yy, pp)["CITL"]))
        else:
            raise ValueError("metric")
    vals = [v for v in vals if np.isfinite(v)]
    if len(vals) == 0:
        return np.nan
    return float(np.var(vals, ddof=1))

def rubin_pool(estimates: List[float], within_vars: List[float], alpha=0.05) -> Dict[str, float]:
    est = np.array([e for e in estimates if np.isfinite(e)], float)
    wv = np.array([v for v in within_vars if np.isfinite(v)], float)
    M = len(est)
    if M == 0:
        return dict(mean=np.nan, se=np.nan, lo=np.nan, hi=np.nan, df=np.nan, M=0)
    if len(wv) != M:
        wv = np.zeros(M, float)

    Qbar = float(np.mean(est))
    W = float(np.mean(wv))
    B = float(np.var(est, ddof=1)) if M > 1 else 0.0
    T = W + (1 + 1 / M) * B
    if T < 0:
        T = 0.0

    df = np.inf
    if B > 0 and W > 0 and M > 1:
        df = (M - 1) * (1 + W / ((1 + 1 / M) * B)) ** 2

    se = float(np.sqrt(T))
    from scipy import stats
    tcrit = float(stats.t.ppf(1 - alpha / 2, df)) if np.isfinite(df) else float(stats.norm.ppf(1 - alpha / 2))
    lo = Qbar - tcrit * se
    hi = Qbar + tcrit * se
    return dict(mean=Qbar, se=se, lo=lo, hi=hi, df=df, M=M)

# ===================== SAVE =====================
def dump_obj(obj, path: Path):
    if not _HAS_JOBLIB:
        logger.info("[SAVE] joblib not installed; skipping model dump.")
        return
    joblib.dump(obj, path)

# ---- NEW: Save SHAP-ready X matrices (exact final pipeline inputs) ----
def _save_shap_X_tree(
    split_name: str,
    task: str,
    model_name: str,
    extras: Dict[str, Any],
    X_before_pre: pd.DataFrame,
    index_values: np.ndarray,
) -> None:
    keep = list(extras.get("final_features", []))
    pre = extras.get("final_preprocessor", None)
    if pre is None or len(keep) == 0:
        raise ValueError(f"SHAP save failed (tree): missing pre/features: {split_name}, {task}, {model_name}")
    X_sel = X_before_pre[keep].copy()
    X_t = pre.transform(X_sel)  # DataFrame float32

    out = X_t.copy()
    out.insert(0, INDEX_COL, np.asarray(index_values))

    p_parq = SHAP_DIR / f"X_{split_name}_{task}_{model_name}.parquet"
    p_csv  = SHAP_DIR / f"X_{split_name}_{task}_{model_name}.csv"
    try:
        out.to_parquet(p_parq, index=False)
    except Exception:
        out.to_csv(p_csv, index=False, encoding="utf-8-sig")

    (SHAP_DIR / f"features_{split_name}_{task}_{model_name}.json").write_text(
        json.dumps({"split": split_name, "task": task, "model": model_name, "features": keep}, ensure_ascii=False, indent=2),
        encoding="utf-8"
    )

def _save_shap_X_lrmi(
    split_name: str,
    task: str,
    mi_tag: str,  # e.g. "LR_MI01"
    extras: Dict[str, Any],
    X_before_pre: pd.DataFrame,
    index_values: np.ndarray,
) -> None:
    keep = list(extras.get("final_features", []))
    pre = extras.get("final_preprocessor", None)
    imp = extras.get("final_imputer", None)
    if pre is None or imp is None or len(keep) == 0:
        raise ValueError(f"SHAP save failed (LR_MI): missing pre/imputer/features: {split_name}, {task}, {mi_tag}")

    X_sel = X_before_pre[keep].copy()
    X_p = pre.transform(X_sel)  # DataFrame with NaN preserved
    X_i = imp.transform(X_p.to_numpy(dtype=np.float32)).astype(np.float32)

    out = pd.DataFrame(X_i, columns=keep)
    out.insert(0, INDEX_COL, np.asarray(index_values))

    p_parq = SHAP_DIR / f"X_{split_name}_{task}_{mi_tag}.parquet"
    p_csv  = SHAP_DIR / f"X_{split_name}_{task}_{mi_tag}.csv"
    try:
        out.to_parquet(p_parq, index=False)
    except Exception:
        out.to_csv(p_csv, index=False, encoding="utf-8-sig")

    (SHAP_DIR / f"features_{split_name}_{task}_{mi_tag}.json").write_text(
        json.dumps({"split": split_name, "task": task, "model": mi_tag, "features": keep}, ensure_ascii=False, indent=2),
        encoding="utf-8"
    )

# ===================== SPLITS =====================
def make_common_split(df: pd.DataFrame, y: np.ndarray):
    sss = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=RANDOM_STATE)
    tr_pos, te_pos = next(sss.split(np.zeros(len(y)), y))
    idx_all = df[INDEX_COL].to_numpy()
    idx_tr, idx_te = idx_all[tr_pos], idx_all[te_pos]
    pd.DataFrame({INDEX_COL: np.r_[idx_tr, idx_te],
                  "Split": (["train"] * len(idx_tr)) + (["test"] * len(idx_te))}).to_csv(
        SPLIT_CSV, index=False, encoding="utf-8-sig"
    )

def make_temporal_split(df_raw: pd.DataFrame):
    if "入院日" not in df_raw.columns:
        return
    admit_dt = pd.to_datetime(df_raw["入院日"], errors="coerce")
    mask_tr_time = (admit_dt >= pd.Timestamp(2018, 4, 1)) & (admit_dt <= pd.Timestamp(2023, 3, 31))
    mask_te_time = (admit_dt >= pd.Timestamp(2023, 4, 1)) & (admit_dt <= pd.Timestamp(2025, 3, 31))

    use_raw = df_raw.drop(columns=[c for c in IGNORE_COLS if c in df_raw.columns], errors="ignore").copy()
    idx_tr_t = use_raw.loc[mask_tr_time, INDEX_COL].to_numpy()
    idx_te_t = use_raw.loc[mask_te_time, INDEX_COL].to_numpy()

    if (len(idx_tr_t) > 0) and (len(idx_te_t) > 0):
        pd.DataFrame({INDEX_COL: np.r_[idx_tr_t, idx_te_t],
                      "SplitTime": (["train_time"] * len(idx_tr_t)) + (["test_time"] * len(idx_te_t))}).to_csv(
            SPLIT_TIME_CSV, index=False, encoding="utf-8-sig"
        )

# ===================== BUILD TRAIN TABLE =====================
def load_and_build_table() -> Tuple[pd.DataFrame, List[str]]:
    raw = pd.read_excel(RAW_PATH)
    logger.info(f"Raw: {raw.shape}")
    use_raw = raw.drop(columns=[c for c in IGNORE_COLS if c in raw.columns], errors="ignore").copy()

    assert INDEX_COL in use_raw.columns, "INDEX required"
    assert TARGET_OUTCOME in use_raw.columns, f"{TARGET_OUTCOME} required"
    assert TARGET_LOS in use_raw.columns, f"{TARGET_LOS} required"
    assert not use_raw[INDEX_COL].duplicated().any(), "INDEX must be unique"

    X0 = build_feature_table(use_raw)
    feat_cols = list(X0.columns)
    base_tbl = pd.concat([use_raw[[INDEX_COL, TARGET_OUTCOME, TARGET_LOS]], X0], axis=1)

    y_all = pd.to_numeric(use_raw[TARGET_OUTCOME], errors="coerce").astype(int).values
    if not SPLIT_CSV.exists():
        make_common_split(use_raw, y_all)
    if not SPLIT_TIME_CSV.exists():
        make_temporal_split(raw)

    return base_tbl, feat_cols

# ===================== FEATURE SETS (TRAIN ONLY) =====================
def compute_or_load_feature_sets(
    X_tr: pd.DataFrame,
    y_tr_out: np.ndarray,
    y_tr_lb: np.ndarray,
    y_tr_los: np.ndarray,
    cache_path: Path,
    feat_cols: List[str],
    force_recompute: bool = False,
) -> Dict[str, List[str]]:
    """
    Compute fixed feature set per task using TRAIN ONLY.
    Cached in cache_path.

    PATCH: validate cache signature (feat_cols + cfg + RAW mtime).
    """
    sig_now = make_cache_signature(feat_cols)

    if (not force_recompute) and cache_path.exists():
        try:
            d = json.loads(cache_path.read_text(encoding="utf-8"))
            if isinstance(d, dict) and ("__signature__" in d) and ("__features__" in d):
                sig_old = d.get("__signature__", {})
                fs_old = d.get("__features__", {})
                if sig_old == sig_now and isinstance(fs_old, dict) and all(isinstance(v, list) for v in fs_old.values()):
                    logger.info(f"[FS] loaded cached (signature match): {cache_path}")
                    return {k: list(v) for k, v in fs_old.items()}
                else:
                    logger.info(f"[FS] cache signature mismatch -> recompute: {cache_path}")
        except Exception:
            logger.info(f"[FS] cache read failed -> recompute: {cache_path}")

    fs = {}
    fs["OUTCOME"] = select_features_once_trainonly(
        X_train=X_tr, y_train=y_tr_out, task="binary",
        topk=_cfg["TOPK_FEATURES"],
        protected=_cfg["PROTECTED_ALWAYS"],
        vif_thr=_cfg["VIF_THR"],
        vif_numeric_only=_cfg["VIF_NUMERIC_ONLY"],
    )
    if ENABLE_LOSBIN:
        fs["LOSBIN"] = select_features_once_trainonly(
            X_train=X_tr, y_train=y_tr_lb, task="binary",
            topk=_cfg["TOPK_FEATURES"],
            protected=_cfg["PROTECTED_ALWAYS"],
            vif_thr=_cfg["VIF_THR"],
            vif_numeric_only=_cfg["VIF_NUMERIC_ONLY"],
        )
    fs["LOSREG"] = select_features_once_trainonly(
        X_train=X_tr, y_train=y_tr_los, task="reg",
        topk=_cfg["TOPK_FEATURES"],
        protected=_cfg["PROTECTED_ALWAYS"],
        vif_thr=_cfg["VIF_THR"],
        vif_numeric_only=_cfg["VIF_NUMERIC_ONLY"],
    )

    payload = {"__signature__": sig_now, "__features__": fs}
    cache_path.write_text(json.dumps(payload, indent=2, ensure_ascii=False), encoding="utf-8")
    logger.info(f"[FS] saved: {cache_path}")
    return fs

# ===================== MAIN: HOLDOUT =====================
def main_holdout():
    tbl, feat_cols = load_and_build_table()
    split = pd.read_csv(SPLIT_CSV)
    t0 = tbl.merge(split, on=INDEX_COL, how="left")

    mask_tr = (t0["Split"] == "train").to_numpy()
    mask_te = (t0["Split"] == "test").to_numpy()

    tr_df = t0.loc[mask_tr].reset_index(drop=True)
    te_df = t0.loc[mask_te].reset_index(drop=True)

    X_tr = tr_df[feat_cols].copy()
    X_te = te_df[feat_cols].copy()

    y_tr_out = tr_df[TARGET_OUTCOME].astype(int).to_numpy()
    y_te_out = te_df[TARGET_OUTCOME].astype(int).to_numpy()

    y_tr_los = tr_df[TARGET_LOS].astype(float).to_numpy()
    y_te_los = te_df[TARGET_LOS].astype(float).to_numpy()

    y_tr_lb = (y_tr_los > LOS_CUTOFF_DAYS).astype(int)
    y_te_lb = (y_te_los > LOS_CUTOFF_DAYS).astype(int)

    # fixed FS per task (holdout TRAIN ONLY)
    fs = compute_or_load_feature_sets(
        X_tr, y_tr_out, y_tr_lb, y_tr_los,
        cache_path=FEATURES_JSON_HOLDOUT,
        feat_cols=feat_cols,
        force_recompute=False,
    )

    # tuning cache
    best_params = {}
    if BESTPARAMS_JSON.exists():
        try:
            best_params = json.loads(BESTPARAMS_JSON.read_text(encoding="utf-8"))
        except Exception:
            best_params = {}

    def _ensure(d, *keys):
        cur = d
        for k in keys:
            if k not in cur:
                cur[k] = {}
            cur = cur[k]
        return d

    # ===================== TUNE (TRAIN ONLY) =====================
    _ensure(best_params, "OUTCOME")
    if "XGB" not in best_params["OUTCOME"] and _HAS_XGB:
        logger.info("[TUNE] OUTCOME XGB (AUROC) ...")
        best_params["OUTCOME"]["XGB"] = tune_xgb_cls_fixedfs(X_tr, y_tr_out, keep=fs["OUTCOME"], metric="AUROC")
    if "LGBM" not in best_params["OUTCOME"]:
        logger.info("[TUNE] OUTCOME LGBM (AUROC) ...")
        best_params["OUTCOME"]["LGBM"] = tune_lgbm_cls_fixedfs(X_tr, y_tr_out, keep=fs["OUTCOME"], metric="AUROC")
    if "LR" not in best_params["OUTCOME"]:
        logger.info("[TUNE] OUTCOME LR (MI leak-safe in CV) ...")
        best_params["OUTCOME"]["LR"] = tune_lr_with_mi_fixedfs(
            X_tr, y_tr_out, keep=fs["OUTCOME"], M=min(5, int(_cfg["MI_M"]))
        )

    if ENABLE_LOSBIN:
        _ensure(best_params, "LOSBIN")
        if "XGB" not in best_params["LOSBIN"] and _HAS_XGB:
            logger.info("[TUNE] LOSBIN XGB (AP) ...")
            best_params["LOSBIN"]["XGB"] = tune_xgb_cls_fixedfs(X_tr, y_tr_lb, keep=fs["LOSBIN"], metric="AP")
        if "LGBM" not in best_params["LOSBIN"]:
            logger.info("[TUNE] LOSBIN LGBM (AP) ...")
            best_params["LOSBIN"]["LGBM"] = tune_lgbm_cls_fixedfs(X_tr, y_tr_lb, keep=fs["LOSBIN"], metric="AP")
        if "LR" not in best_params["LOSBIN"]:
            logger.info("[TUNE] LOSBIN LR (MI leak-safe in CV) ...")
            best_params["LOSBIN"]["LR"] = tune_lr_with_mi_fixedfs(
                X_tr, y_tr_lb, keep=fs["LOSBIN"], M=min(5, int(_cfg["MI_M"]))
            )

    _ensure(best_params, "LOSREG")
    if "XGBReg" not in best_params["LOSREG"] and _HAS_XGB:
        logger.info("[TUNE] LOSREG XGB reg ...")
        best_params["LOSREG"]["XGBReg"] = tune_xgb_reg_fixedfs(X_tr, y_tr_los, keep=fs["LOSREG"])
    if "LGBMReg" not in best_params["LOSREG"]:
        logger.info("[TUNE] LOSREG LGBM regression ...")
        best_params["LOSREG"]["LGBMReg"] = tune_lgbm_reg_fixedfs(X_tr, y_tr_los, keep=fs["LOSREG"], objective_name="regression")
    if "LGBMRegL1" not in best_params["LOSREG"]:
        logger.info("[TUNE] LOSREG LGBM regression_l1 ...")
        best_params["LOSREG"]["LGBMRegL1"] = tune_lgbm_reg_fixedfs(X_tr, y_tr_los, keep=fs["LOSREG"], objective_name="regression_l1")

    BESTPARAMS_JSON.write_text(json.dumps(best_params, indent=2, ensure_ascii=False), encoding="utf-8")
    logger.info(f"Saved best params: {BESTPARAMS_JSON}")

    # ===================== FIT/EVAL ON HOLDOUT =====================
    rec_cls_rows, pooled_rows, rec_reg_rows = [], [], []
    cls_metrics = ["AUROC", "AP", "Brier", "CalibSlope", "CITL"]
    M = int(_cfg["MI_M"])

    # ---- OUTCOME trees ----
    for mdl in (["XGB"] if _HAS_XGB else []) + ["LGBM"]:
        oof_cal, te_cal, cal, extras = oof_and_test_tree_classifier_fixedfs(
            Xtr=X_tr, ytr=y_tr_out, Xte=X_te, yte=y_te_out,
            keep=fs["OUTCOME"], model_name=mdl, best_params=best_params["OUTCOME"][mdl],
        )
        m = safe_cls_metrics(y_te_out, te_cal)
        rec_cls_rows.append({"Task": "OUTCOME", "Model": mdl, "Split": "Holdout", "N": int(len(y_te_out)), **m})

        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_out,
            "p_cal": te_cal.astype(np.float32),
        }).to_csv(PRED_DIR / f"OUTCOME_{mdl}_Holdout.csv", index=False, encoding="utf-8-sig")

        if _HAS_JOBLIB:
            dump_obj(extras, MODEL_DIR / f"OUTCOME_{mdl}_Holdout_final.joblib")

        # ---- NEW: SHAP X save ----
        _save_shap_X_tree(
            split_name="Holdout",
            task="OUTCOME",
            model_name=mdl,
            extras=extras,
            X_before_pre=X_te,
            index_values=te_df[INDEX_COL].to_numpy(),
        )

    # ---- OUTCOME LR MI pooled ----
    lr_oof_list, lr_te_list, lr_extras_list = oof_and_test_lr_with_mi_fixedfs(
        Xtr=X_tr, ytr=y_tr_out, Xte=X_te, yte=y_te_out,
        keep=fs["OUTCOME"], best_params=best_params["OUTCOME"]["LR"], M=M,
    )
    est_map = {k: [] for k in cls_metrics}
    var_map = {k: [] for k in cls_metrics}

    for i in range(M):
        p = np.asarray(lr_te_list[i], float)
        m_i = safe_cls_metrics(y_te_out, p)
        rec_cls_rows.append({"Task": "OUTCOME", "Model": "LR_MI", "Split": "Holdout", "MI": i+1, "N": int(len(y_te_out)), **m_i})

        for k in cls_metrics:
            est_map[k].append(m_i[k])
            var_map[k].append(paired_boot_var_cls(y_te_out, p, metric=k, B=_cfg["BOOT_B_CLASS"], seed=1000 + i))

        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_out,
            "p_cal": p.astype(np.float32),
        }).to_csv(PRED_DIR / f"OUTCOME_LR_MI{(i+1):02d}_Holdout.csv", index=False, encoding="utf-8-sig")

        if _HAS_JOBLIB:
            dump_obj(lr_extras_list[i], MODEL_DIR / f"OUTCOME_LR_MI{(i+1):02d}_Holdout_final.joblib")

        # ---- NEW: SHAP X save (LR_MI each MI) ----
        _save_shap_X_lrmi(
            split_name="Holdout",
            task="OUTCOME",
            mi_tag=f"LR_MI{(i+1):02d}",
            extras=lr_extras_list[i],
            X_before_pre=X_te,
            index_values=te_df[INDEX_COL].to_numpy(),
        )

    for k in cls_metrics:
        pool = rubin_pool(est_map[k], var_map[k])
        pooled_rows.append({"Task": "OUTCOME", "Model": "LR_MI", "Split": "Holdout", "Metric": k, **pool})

    # ---- LOSBIN ----
    if ENABLE_LOSBIN:
        for mdl in (["XGB"] if _HAS_XGB else []) + ["LGBM"]:
            oof_cal, te_cal, cal, extras = oof_and_test_tree_classifier_fixedfs(
                Xtr=X_tr, ytr=y_tr_lb, Xte=X_te, yte=y_te_lb,
                keep=fs["LOSBIN"], model_name=mdl, best_params=best_params["LOSBIN"][mdl],
            )
            m = safe_cls_metrics(y_te_lb, te_cal)
            rec_cls_rows.append({"Task": "LOSBIN", "Model": mdl, "Split": "Holdout", "N": int(len(y_te_lb)), **m})

            pd.DataFrame({
                INDEX_COL: te_df[INDEX_COL].to_numpy(),
                "y_true": y_te_lb,
                "p_cal": te_cal.astype(np.float32),
            }).to_csv(PRED_DIR / f"LOSBIN_{mdl}_Holdout.csv", index=False, encoding="utf-8-sig")

            if _HAS_JOBLIB:
                dump_obj(extras, MODEL_DIR / f"LOSBIN_{mdl}_Holdout_final.joblib")

            # ---- NEW: SHAP X save ----
            _save_shap_X_tree(
                split_name="Holdout",
                task="LOSBIN",
                model_name=mdl,
                extras=extras,
                X_before_pre=X_te,
                index_values=te_df[INDEX_COL].to_numpy(),
            )

        lr_oof_list, lr_te_list, lr_extras_list = oof_and_test_lr_with_mi_fixedfs(
            Xtr=X_tr, ytr=y_tr_lb, Xte=X_te, yte=y_te_lb,
            keep=fs["LOSBIN"], best_params=best_params["LOSBIN"]["LR"], M=M,
        )
        est_map = {k: [] for k in cls_metrics}
        var_map = {k: [] for k in cls_metrics}

        for i in range(M):
            p = np.asarray(lr_te_list[i], float)
            m_i = safe_cls_metrics(y_te_lb, p)
            rec_cls_rows.append({"Task": "LOSBIN", "Model": "LR_MI", "Split": "Holdout", "MI": i+1, "N": int(len(y_te_lb)), **m_i})

            for k in cls_metrics:
                est_map[k].append(m_i[k])
                var_map[k].append(paired_boot_var_cls(y_te_lb, p, metric=k, B=_cfg["BOOT_B_CLASS"], seed=2000 + i))

            pd.DataFrame({
                INDEX_COL: te_df[INDEX_COL].to_numpy(),
                "y_true": y_te_lb,
                "p_cal": p.astype(np.float32),
            }).to_csv(PRED_DIR / f"LOSBIN_LR_MI{(i+1):02d}_Holdout.csv", index=False, encoding="utf-8-sig")

            if _HAS_JOBLIB:
                dump_obj(lr_extras_list[i], MODEL_DIR / f"LOSBIN_LR_MI{(i+1):02d}_Holdout_final.joblib")

            # ---- NEW: SHAP X save (LR_MI each MI) ----
            _save_shap_X_lrmi(
                split_name="Holdout",
                task="LOSBIN",
                mi_tag=f"LR_MI{(i+1):02d}",
                extras=lr_extras_list[i],
                X_before_pre=X_te,
                index_values=te_df[INDEX_COL].to_numpy(),
            )

        for k in cls_metrics:
            pool = rubin_pool(est_map[k], var_map[k])
            pooled_rows.append({"Task": "LOSBIN", "Model": "LR_MI", "Split": "Holdout", "Metric": k, **pool})

    # ---- LOSREG (trees) ----
    if _HAS_XGB:
        oof_r, pred_te, extras = oof_and_test_tree_regressor_fixedfs(
            Xtr=X_tr, ytr=y_tr_los, Xte=X_te, yte=y_te_los,
            keep=fs["LOSREG"], model_name="XGB", best_params=best_params["LOSREG"]["XGBReg"],
            reg_objective="regression",
        )
        rec_reg_rows.append({
            "Task": "LOSREG", "Model": "XGBReg", "Split": "Holdout", "N": int(len(y_te_los)),
            "RMSE": rmse(y_te_los, pred_te),
            "MAE": float(np.mean(np.abs(y_te_los - pred_te))),
            "MedAE": float(median_absolute_error(y_te_los, pred_te)),
            "R2": float(r2_score(y_te_los, pred_te)) if len(y_te_los) >= 2 else np.nan,
            "SMAPE": smape(y_te_los, pred_te),
            "SmearFull": float(extras.get("smear_full", np.nan)),
        })
        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_los.astype(np.float32),
            "y_pred": pred_te.astype(np.float32),
            "Events": y_te_out.astype(int),
        }).to_csv(PRED_DIR / "LOSREG_XGBReg_Holdout.csv", index=False, encoding="utf-8-sig")
        if _HAS_JOBLIB:
            dump_obj(extras, MODEL_DIR / "LOSREG_XGBReg_Holdout_final.joblib")

    for key, obj in [("LGBMReg", "regression"), ("LGBMRegL1", "regression_l1")]:
        oof_r, pred_te, extras = oof_and_test_tree_regressor_fixedfs(
            Xtr=X_tr, ytr=y_tr_los, Xte=X_te, yte=y_te_los,
            keep=fs["LOSREG"], model_name="LGBM", best_params=best_params["LOSREG"][key],
            reg_objective=obj,
        )
        rec_reg_rows.append({
            "Task": "LOSREG", "Model": key, "Split": "Holdout", "N": int(len(y_te_los)),
            "RMSE": rmse(y_te_los, pred_te),
            "MAE": float(np.mean(np.abs(y_te_los - pred_te))),
            "MedAE": float(median_absolute_error(y_te_los, pred_te)),
            "R2": float(r2_score(y_te_los, pred_te)) if len(y_te_los) >= 2 else np.nan,
            "SMAPE": smape(y_te_los, pred_te),
            "SmearFull": float(extras.get("smear_full", np.nan)),
        })
        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_los.astype(np.float32),
            "y_pred": pred_te.astype(np.float32),
            "Events": y_te_out.astype(int),
        }).to_csv(PRED_DIR / f"LOSREG_{key}_Holdout.csv", index=False, encoding="utf-8-sig")
        if _HAS_JOBLIB:
            dump_obj(extras, MODEL_DIR / f"LOSREG_{key}_Holdout_final.joblib")

    with pd.ExcelWriter(EVAL_XLSX) as w:
        pd.DataFrame(rec_cls_rows).to_excel(w, sheet_name="Classification(Holdout)", index=False)
        pd.DataFrame(rec_reg_rows).to_excel(w, sheet_name="Regression(Holdout)", index=False)
        if pooled_rows:
            pd.DataFrame(pooled_rows).to_excel(w, sheet_name="MI_Pooled(LR)", index=False)

    logger.info(f"Finished HOLDOUT. Outputs: {RUN_DIR.resolve()}")

# ===================== TEMPORAL VALIDATION =====================
def temporal_validation():
    if not SPLIT_TIME_CSV.exists():
        raise RuntimeError("Temporal split file not found. Ensure 入院日 exists and holdout created it.")
    if not BESTPARAMS_JSON.exists():
        raise RuntimeError("best_params.json not found. Run holdout first.")

    tbl, feat_cols = load_and_build_table()
    split_t = pd.read_csv(SPLIT_TIME_CSV)
    t0 = tbl.merge(split_t, on=INDEX_COL, how="left")

    tr = (t0["SplitTime"] == "train_time").to_numpy()
    te = (t0["SplitTime"] == "test_time").to_numpy()

    tr_df = t0.loc[tr].reset_index(drop=True)
    te_df = t0.loc[te].reset_index(drop=True)

    X_tr = tr_df[feat_cols].copy()
    X_te = te_df[feat_cols].copy()

    y_tr_out = tr_df[TARGET_OUTCOME].astype(int).to_numpy()
    y_te_out = te_df[TARGET_OUTCOME].astype(int).to_numpy()

    y_tr_los = tr_df[TARGET_LOS].astype(float).to_numpy()
    y_te_los = te_df[TARGET_LOS].astype(float).to_numpy()

    y_tr_lb = (y_tr_los > LOS_CUTOFF_DAYS).astype(int)
    y_te_lb = (y_te_los > LOS_CUTOFF_DAYS).astype(int)

    best_params = json.loads(BESTPARAMS_JSON.read_text(encoding="utf-8"))

    # temporal FS computed on temporal TRAIN ONLY, using separate cache file
    fs_temp = compute_or_load_feature_sets(
        X_tr, y_tr_out, y_tr_lb, y_tr_los,
        cache_path=FEATURES_JSON_TEMPORAL,
        feat_cols=feat_cols,
        force_recompute=False,
    )

    rows_cls, rows_reg, rows_pool = [], [], []
    cls_metrics = ["AUROC", "AP", "Brier", "CalibSlope", "CITL"]
    M = int(_cfg["MI_M"])

    for mdl in (["XGB"] if _HAS_XGB and "XGB" in best_params["OUTCOME"] else []) + ["LGBM"]:
        oof_cal, te_cal, cal, extras = oof_and_test_tree_classifier_fixedfs(
            Xtr=X_tr, ytr=y_tr_out, Xte=X_te, yte=y_te_out,
            keep=fs_temp["OUTCOME"], model_name=mdl, best_params=best_params["OUTCOME"][mdl],
        )
        m = safe_cls_metrics(y_te_out, te_cal)
        rows_cls.append({"Task": "OUTCOME", "Model": mdl, "Split": "Temporal", "N": int(len(y_te_out)), **m})
        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_out,
            "p_cal": te_cal.astype(np.float32),
        }).to_csv(PRED_DIR / f"Temporal_OUTCOME_{mdl}.csv", index=False, encoding="utf-8-sig")
        if _HAS_JOBLIB:
            dump_obj(extras, MODEL_DIR / f"Temporal_OUTCOME_{mdl}_final.joblib")

        # ---- NEW: SHAP X save ----
        _save_shap_X_tree(
            split_name="Temporal",
            task="OUTCOME",
            model_name=mdl,
            extras=extras,
            X_before_pre=X_te,
            index_values=te_df[INDEX_COL].to_numpy(),
        )

    lr_oof_list, lr_te_list, lr_extras_list = oof_and_test_lr_with_mi_fixedfs(
        Xtr=X_tr, ytr=y_tr_out, Xte=X_te, yte=y_te_out,
        keep=fs_temp["OUTCOME"], best_params=best_params["OUTCOME"]["LR"], M=M,
    )
    est_map = {k: [] for k in cls_metrics}
    var_map = {k: [] for k in cls_metrics}

    for i in range(M):
        p = np.asarray(lr_te_list[i], float)
        m_i = safe_cls_metrics(y_te_out, p)
        rows_cls.append({"Task": "OUTCOME", "Model": "LR_MI", "Split": "Temporal", "MI": i+1, "N": int(len(y_te_out)), **m_i})
        for k in cls_metrics:
            est_map[k].append(m_i[k])
            var_map[k].append(paired_boot_var_cls(y_te_out, p, metric=k, B=_cfg["BOOT_B_CLASS"], seed=7000 + i))
        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_out,
            "p_cal": p.astype(np.float32),
        }).to_csv(PRED_DIR / f"Temporal_OUTCOME_LR_MI{(i+1):02d}.csv", index=False, encoding="utf-8-sig")
        if _HAS_JOBLIB:
            dump_obj(lr_extras_list[i], MODEL_DIR / f"Temporal_OUTCOME_LR_MI{(i+1):02d}_final.joblib")

        # ---- NEW: SHAP X save ----
        _save_shap_X_lrmi(
            split_name="Temporal",
            task="OUTCOME",
            mi_tag=f"LR_MI{(i+1):02d}",
            extras=lr_extras_list[i],
            X_before_pre=X_te,
            index_values=te_df[INDEX_COL].to_numpy(),
        )

    for k in cls_metrics:
        pool = rubin_pool(est_map[k], var_map[k])
        rows_pool.append({"Task": "OUTCOME", "Model": "LR_MI", "Split": "Temporal", "Metric": k, **pool})

    if ENABLE_LOSBIN:
        for mdl in (["XGB"] if _HAS_XGB and "XGB" in best_params["LOSBIN"] else []) + ["LGBM"]:
            oof_cal, te_cal, cal, extras = oof_and_test_tree_classifier_fixedfs(
                Xtr=X_tr, ytr=y_tr_lb, Xte=X_te, yte=y_te_lb,
                keep=fs_temp["LOSBIN"], model_name=mdl, best_params=best_params["LOSBIN"][mdl],
            )
            m = safe_cls_metrics(y_te_lb, te_cal)
            rows_cls.append({"Task": "LOSBIN", "Model": mdl, "Split": "Temporal", "N": int(len(y_te_lb)), **m})
            pd.DataFrame({
                INDEX_COL: te_df[INDEX_COL].to_numpy(),
                "y_true": y_te_lb,
                "p_cal": te_cal.astype(np.float32),
            }).to_csv(PRED_DIR / f"Temporal_LOSBIN_{mdl}.csv", index=False, encoding="utf-8-sig")
            if _HAS_JOBLIB:
                dump_obj(extras, MODEL_DIR / f"Temporal_LOSBIN_{mdl}_final.joblib")

            # ---- NEW: SHAP X save ----
            _save_shap_X_tree(
                split_name="Temporal",
                task="LOSBIN",
                model_name=mdl,
                extras=extras,
                X_before_pre=X_te,
                index_values=te_df[INDEX_COL].to_numpy(),
            )

        lr_oof_list, lr_te_list, lr_extras_list = oof_and_test_lr_with_mi_fixedfs(
            Xtr=X_tr, ytr=y_tr_lb, Xte=X_te, yte=y_te_lb,
            keep=fs_temp["LOSBIN"], best_params=best_params["LOSBIN"]["LR"], M=M,
        )
        est_map = {k: [] for k in cls_metrics}
        var_map = {k: [] for k in cls_metrics}

        for i in range(M):
            p = np.asarray(lr_te_list[i], float)
            m_i = safe_cls_metrics(y_te_lb, p)
            rows_cls.append({"Task": "LOSBIN", "Model": "LR_MI", "Split": "Temporal", "MI": i+1, "N": int(len(y_te_lb)), **m_i})
            for k in cls_metrics:
                est_map[k].append(m_i[k])
                var_map[k].append(paired_boot_var_cls(y_te_lb, p, metric=k, B=_cfg["BOOT_B_CLASS"], seed=8000 + i))
            pd.DataFrame({
                INDEX_COL: te_df[INDEX_COL].to_numpy(),
                "y_true": y_te_lb,
                "p_cal": p.astype(np.float32),
            }).to_csv(PRED_DIR / f"Temporal_LOSBIN_LR_MI{(i+1):02d}.csv", index=False, encoding="utf-8-sig")
            if _HAS_JOBLIB:
                dump_obj(lr_extras_list[i], MODEL_DIR / f"Temporal_LOSBIN_LR_MI{(i+1):02d}_final.joblib")

            # ---- NEW: SHAP X save ----
            _save_shap_X_lrmi(
                split_name="Temporal",
                task="LOSBIN",
                mi_tag=f"LR_MI{(i+1):02d}",
                extras=lr_extras_list[i],
                X_before_pre=X_te,
                index_values=te_df[INDEX_COL].to_numpy(),
            )

        for k in cls_metrics:
            pool = rubin_pool(est_map[k], var_map[k])
            rows_pool.append({"Task": "LOSBIN", "Model": "LR_MI", "Split": "Temporal", "Metric": k, **pool})

    if _HAS_XGB and "XGBReg" in best_params["LOSREG"]:
        oof_r, pred_te, extras = oof_and_test_tree_regressor_fixedfs(
            Xtr=X_tr, ytr=y_tr_los, Xte=X_te, yte=y_te_los,
            keep=fs_temp["LOSREG"], model_name="XGB", best_params=best_params["LOSREG"]["XGBReg"],
            reg_objective="regression",
        )
        rows_reg.append({
            "Task": "LOSREG", "Model": "XGBReg", "Split": "Temporal", "N": int(len(y_te_los)),
            "RMSE": rmse(y_te_los, pred_te),
            "MAE": float(np.mean(np.abs(y_te_los - pred_te))),
            "MedAE": float(median_absolute_error(y_te_los, pred_te)),
            "R2": float(r2_score(y_te_los, pred_te)) if len(y_te_los) >= 2 else np.nan,
            "SMAPE": smape(y_te_los, pred_te),
            "SmearFull": float(extras.get("smear_full", np.nan)),
        })
        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_los.astype(np.float32),
            "y_pred": pred_te.astype(np.float32),
            "Events": y_te_out.astype(int),
        }).to_csv(PRED_DIR / "Temporal_LOSREG_XGBReg.csv", index=False, encoding="utf-8-sig")
        if _HAS_JOBLIB:
            dump_obj(extras, MODEL_DIR / "Temporal_LOSREG_XGBReg_final.joblib")

    for key, obj in [("LGBMReg", "regression"), ("LGBMRegL1", "regression_l1")]:
        oof_r, pred_te, extras = oof_and_test_tree_regressor_fixedfs(
            Xtr=X_tr, ytr=y_tr_los, Xte=X_te, yte=y_te_los,
            keep=fs_temp["LOSREG"], model_name="LGBM", best_params=best_params["LOSREG"][key],
            reg_objective=obj,
        )
        rows_reg.append({
            "Task": "LOSREG", "Model": key, "Split": "Temporal", "N": int(len(y_te_los)),
            "RMSE": rmse(y_te_los, pred_te),
            "MAE": float(np.mean(np.abs(y_te_los - pred_te))),
            "MedAE": float(median_absolute_error(y_te_los, pred_te)),
            "R2": float(r2_score(y_te_los, pred_te)) if len(y_te_los) >= 2 else np.nan,
            "SMAPE": smape(y_te_los, pred_te),
            "SmearFull": float(extras.get("smear_full", np.nan)),
        })
        pd.DataFrame({
            INDEX_COL: te_df[INDEX_COL].to_numpy(),
            "y_true": y_te_los.astype(np.float32),
            "y_pred": pred_te.astype(np.float32),
            "Events": y_te_out.astype(int),
        }).to_csv(PRED_DIR / f"Temporal_LOSREG_{key}.csv", index=False, encoding="utf-8-sig")
        if _HAS_JOBLIB:
            dump_obj(extras, MODEL_DIR / f"Temporal_LOSREG_{key}_final.joblib")

    with pd.ExcelWriter(EVAL_TEMP_XLSX) as w:
        pd.DataFrame(rows_cls).to_excel(w, sheet_name="Classification(Temporal)", index=False)
        pd.DataFrame(rows_reg).to_excel(w, sheet_name="Regression(Temporal)", index=False)
        if rows_pool:
            pd.DataFrame(rows_pool).to_excel(w, sheet_name="MI_Pooled(LR_Temporal)", index=False)
        pd.DataFrame({
            "Train_N": [len(y_tr_out)], "Train_PosRate": [float(np.mean(y_tr_out))],
            "Test_N": [len(y_te_out)], "Test_PosRate": [float(np.mean(y_te_out))],
        }).to_excel(w, sheet_name="Temporal_Descriptives", index=False)

    logger.info(f"Finished TEMPORAL. Outputs: {RUN_DIR.resolve()}")

# ===================== CLI =====================
def _parse_cli():
    parser = argparse.ArgumentParser(description="Stroke modeling pipeline (NO FIGURES)", add_help=True)
    parser.add_argument("--run", choices=["holdout", "temporal", "both"], default="both")
    parser.add_argument("--mode", choices=["MAX", "SPEED"], default="MAX")
    parser.add_argument("--mi", type=int, default=None, help="Override MI_M (LR only), e.g., 5 or 10")
    in_notebook = "ipykernel" in sys.modules
    if in_notebook:
        args, _ = parser.parse_known_args([])
    else:
        args, _ = parser.parse_known_args()
    return args

if __name__ == "__main__":
    args = _parse_cli()

    if args.mode and args.mode.upper() != MODE:
        MODE = args.mode.upper()
        _cfg = _cfg_by_mode(MODE)
        save_env_and_config()

    if args.mi is not None:
        _cfg["MI_M"] = int(args.mi)
        save_env_and_config()

    try:
        if args.run in ("holdout", "both"):
            main_holdout()
        if args.run in ("temporal", "both"):
            temporal_validation()
    except Exception as e:
        logger.exception(f"Unhandled error: {e}")
        raise

# ===== END OF FILE =====
```

```python
# -*- coding: utf-8 -*-
"""
Make paper-ready Tables/Figures from the DEV pipeline outputs (CSV-based).
This script matches the exact filenames/columns produced by your pipeline:

Classification predictions:
  - Holdout: OUTCOME_{XGB|LGBM}_Holdout.csv, OUTCOME_LR_MI01_Holdout.csv ... (INDEX,y_true,p_cal)
  - Holdout: LOSBIN_{XGB|LGBM}_Holdout.csv, LOSBIN_LR_MI01_Holdout.csv ... (INDEX,y_true,p_cal)
  - Temporal: Temporal_OUTCOME_{XGB|LGBM}.csv, Temporal_OUTCOME_LR_MI01.csv ... (INDEX,y_true,p_cal)
  - Temporal: Temporal_LOSBIN_{XGB|LGBM}.csv, Temporal_LOSBIN_LR_MI01.csv ... (INDEX,y_true,p_cal)

Regression predictions (continuous LOS):
  - Holdout: LOSREG_XGBReg_Holdout.csv, LOSREG_LGBMRegL1_Holdout.csv (INDEX,y_true,y_pred,Events)
  - Temporal: Temporal_LOSREG_XGBReg.csv, Temporal_LOSREG_LGBMRegL1.csv (INDEX,y_true,y_pred,Events)

User requirement:
  - Continuous (LOSREG) uses only XGBReg + LGBMRegL1 (display name: LightGBM).
"""

from __future__ import annotations
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import roc_auc_score, average_precision_score

# ------------------- CONFIG -------------------
RANDOM_STATE = 42
RUN_DIR = Path(f"results_paper_ready_seed{RANDOM_STATE}")
PRED_DIR = RUN_DIR / "predictions"
OUT_DIR = RUN_DIR / "paper_outputs"
OUT_DIR.mkdir(parents=True, exist_ok=True)

LOS_CUTOFF_DAYS = 21

# classification models to include (if files exist)
CLS_MODELS_TREE = ["XGB", "LGBM"]  # will be auto-skipped if missing
INCLUDE_LR_MI = True              # pooled by MI-average p_cal

# continuous LOSREG models (strict requirement by user)
REG_MODELS = [
    ("XGBReg", "XGBoost"),
    ("LGBMRegL1", "LightGBM"),  # display as LGBM
]

# ------------------- UTIL -------------------
def _read_csv(path: Path) -> pd.DataFrame:
    df = pd.read_csv(path)
    return df

def _find_mi_files(prefix: str, split_tag: str) -> list[Path]:
    """
    Example:
      prefix='OUTCOME_LR_MI', split_tag='Holdout' -> OUTCOME_LR_MI01_Holdout.csv ...
      prefix='Temporal_OUTCOME_LR_MI', split_tag='' -> Temporal_OUTCOME_LR_MI01.csv ...
    """
    files = []
    if split_tag:
        pat = re.compile(rf"^{re.escape(prefix)}(\d{{2}})_{re.escape(split_tag)}\.csv$", re.IGNORECASE)
    else:
        pat = re.compile(rf"^{re.escape(prefix)}(\d{{2}})\.csv$", re.IGNORECASE)

    for p in PRED_DIR.glob("*.csv"):
        m = pat.match(p.name)
        if m:
            files.append((int(m.group(1)), p))
    files = [p for _, p in sorted(files, key=lambda x: x[0])]
    return files

def _load_cls(split: str, task: str, model: str) -> pd.DataFrame:
    """
    split: 'Holdout' or 'Temporal'
    task: 'OUTCOME' or 'LOSBIN'
    model:
      - 'XGB'/'LGBM' trees -> single file
      - 'LR_MI' -> MI-average across multiple files
    returns df with columns: INDEX,y_true,p_cal
    """
    if split == "Holdout":
        if model in ["XGB", "LGBM"]:
            f = PRED_DIR / f"{task}_{model}_Holdout.csv"
            if not f.exists():
                raise FileNotFoundError(f"Missing: {f}")
            df = _read_csv(f)
            need = {"INDEX", "y_true", "p_cal"}
            if not need.issubset(df.columns):
                raise ValueError(f"{f.name} columns={list(df.columns)} required={sorted(need)}")
            return df[["INDEX","y_true","p_cal"]].copy()

        if model == "LR_MI":
            files = _find_mi_files(f"{task}_LR_MI", "Holdout")
            if len(files) == 0:
                raise FileNotFoundError(f"No MI files for {task} Holdout (expected {task}_LR_MI01_Holdout.csv ...)")
            dfs = [_read_csv(p) for p in files]
            for d, p in zip(dfs, files):
                need = {"INDEX","y_true","p_cal"}
                if not need.issubset(d.columns):
                    raise ValueError(f"{p.name} columns={list(d.columns)} required={sorted(need)}")

            base = dfs[0][["INDEX","y_true"]].copy()
            ps = []
            for d in dfs:
                dd = d[["INDEX","p_cal"]].copy()
                base = base.merge(dd, on="INDEX", how="inner", suffixes=("","_x"))
                ps.append(d.set_index("INDEX")["p_cal"])
            # MI-average by INDEX
            pmat = pd.concat(ps, axis=1).loc[base["INDEX"]].to_numpy(dtype=float)
            out = base[["INDEX","y_true"]].copy()
            out["p_cal"] = np.mean(pmat, axis=1).astype(np.float32)
            return out

    if split == "Temporal":
        if model in ["XGB", "LGBM"]:
            f = PRED_DIR / f"Temporal_{task}_{model}.csv"
            if not f.exists():
                raise FileNotFoundError(f"Missing: {f}")
            df = _read_csv(f)
            need = {"INDEX","y_true","p_cal"}
            if not need.issubset(df.columns):
                raise ValueError(f"{f.name} columns={list(df.columns)} required={sorted(need)}")
            return df[["INDEX","y_true","p_cal"]].copy()

        if model == "LR_MI":
            files = _find_mi_files(f"Temporal_{task}_LR_MI", "")
            if len(files) == 0:
                raise FileNotFoundError(f"No MI files for Temporal {task} (expected Temporal_{task}_LR_MI01.csv ...)")
            dfs = [_read_csv(p) for p in files]
            for d, p in zip(dfs, files):
                need = {"INDEX","y_true","p_cal"}
                if not need.issubset(d.columns):
                    raise ValueError(f"{p.name} columns={list(d.columns)} required={sorted(need)}")
            base = dfs[0][["INDEX","y_true"]].copy()
            ps = [d.set_index("INDEX")["p_cal"] for d in dfs]
            pmat = pd.concat(ps, axis=1).loc[base["INDEX"]].to_numpy(dtype=float)
            out = base.copy()
            out["p_cal"] = np.mean(pmat, axis=1).astype(np.float32)
            return out

    raise ValueError("split must be Holdout/Temporal")

def _load_reg(split: str, model_key: str) -> pd.DataFrame:
    """
    returns df with columns: INDEX,y_true,y_pred,Events
    """
    if split == "Holdout":
        f = PRED_DIR / f"LOSREG_{model_key}_Holdout.csv"
        if model_key == "XGBReg":
            # XGB uses fixed filename in your code
            f2 = PRED_DIR / "LOSREG_XGBReg_Holdout.csv"
            if f2.exists():
                f = f2
        if not f.exists():
            raise FileNotFoundError(f"Missing: {f}")
    else:
        f = PRED_DIR / f"Temporal_LOSREG_{model_key}.csv"
        if model_key == "XGBReg":
            f2 = PRED_DIR / "Temporal_LOSREG_XGBReg.csv"
            if f2.exists():
                f = f2
        if not f.exists():
            raise FileNotFoundError(f"Missing: {f}")

    df = _read_csv(f)
    need = {"INDEX","y_true","y_pred"}
    if not need.issubset(df.columns):
        raise ValueError(f"{f.name} columns={list(df.columns)} required={sorted(need)}")
    # Eventsはあってもなくても動く
    if "Events" not in df.columns:
        df["Events"] = np.nan
    return df[["INDEX","y_true","y_pred","Events"]].copy()

def _cls_metrics(y: np.ndarray, p: np.ndarray) -> dict:
    y = np.asarray(y, int)
    p = np.clip(np.asarray(p, float), 1e-6, 1-1e-6)
    out = dict(AUROC=np.nan, AUPRC=np.nan, Brier=float(np.mean((p-y)**2)))
    if np.unique(y).size >= 2:
        out["AUROC"] = float(roc_auc_score(y, p))
        out["AUPRC"] = float(average_precision_score(y, p))
    return out

def _reg_metrics(y: np.ndarray, yhat: np.ndarray) -> dict:
    y = np.asarray(y, float)
    yhat = np.asarray(yhat, float)
    rmse = float(np.sqrt(np.mean((y-yhat)**2)))
    mae = float(np.mean(np.abs(y-yhat)))
    med = float(np.median(np.abs(y-yhat)))
    # R2 manually (avoid sklearn import)
    ss_res = float(np.sum((y-yhat)**2))
    ss_tot = float(np.sum((y-np.mean(y))**2))
    r2 = float(1 - ss_res/ss_tot) if ss_tot > 0 else np.nan
    denom = np.abs(y)+np.abs(yhat)
    denom = np.where(denom==0, 1.0, denom)
    smape = float(np.mean(2*np.abs(yhat-y)/denom))
    return dict(RMSE=rmse, MAE=mae, MedAE=med, R2=r2, SMAPE=smape)

def _calibration_curve_equal_width(y, p, n_bins=10):
    y = np.asarray(y, int)
    p = np.asarray(p, float)
    bins = np.linspace(0, 1, n_bins+1)
    ids = np.digitize(p, bins) - 1
    xs, ys, ns = [], [], []
    for b in range(n_bins):
        m = ids == b
        if np.sum(m) == 0:
            continue
        xs.append(float(np.mean(p[m])))
        ys.append(float(np.mean(y[m])))
        ns.append(int(np.sum(m)))
    return np.array(xs), np.array(ys), np.array(ns)

def _inventory():
    csvs = sorted([p.name for p in PRED_DIR.glob("*.csv")])
    (OUT_DIR / "inventory_predictions.txt").write_text("\n".join(csvs), encoding="utf-8")
    print("[OK] inventory written:", OUT_DIR / "inventory_predictions.txt")
    return csvs

# ------------------- TABLES -------------------
def make_table_metrics(split: str) -> pd.DataFrame:
    rows = []

    # Classification tasks
    for task in ["OUTCOME", "LOSBIN"]:
        # trees
        for mdl in CLS_MODELS_TREE:
            try:
                df = _load_cls(split, task, mdl)
            except FileNotFoundError:
                continue
            m = _cls_metrics(df["y_true"].to_numpy(), df["p_cal"].to_numpy())
            rows.append(dict(Split=split, Task=task, Model=mdl, N=len(df), **m))

        # LR MI average
        if INCLUDE_LR_MI:
            try:
                df = _load_cls(split, task, "LR_MI")
                m = _cls_metrics(df["y_true"].to_numpy(), df["p_cal"].to_numpy())
                rows.append(dict(Split=split, Task=task, Model="LR_MI", N=len(df), **m))
            except FileNotFoundError:
                pass

    # Regression (continuous LOS) — only required two models
    for key, disp in REG_MODELS:
        df = _load_reg(split, key)
        m = _reg_metrics(df["y_true"].to_numpy(), df["y_pred"].to_numpy())
        rows.append(dict(Split=split, Task="LOSREG", Model=disp, N=len(df), **m))

    out = pd.DataFrame(rows)
    out.to_excel(OUT_DIR / f"Table_metrics_{split}.xlsx", index=False)
    print("[OK] saved:", OUT_DIR / f"Table_metrics_{split}.xlsx")
    return out

# ------------------- FIGURES -------------------
def fig_calibration(split: str, task: str, models: list[str], fname: str):
    plt.figure()
    # perfect line
    plt.plot([0,1],[0,1])
    for mdl in models:
        try:
            df = _load_cls(split, task, mdl)
        except FileNotFoundError:
            continue
        x, y, n = _calibration_curve_equal_width(df["y_true"].to_numpy(), df["p_cal"].to_numpy(), n_bins=10)
        plt.plot(x, y)
    plt.xlabel("Mean predicted risk")
    plt.ylabel("Observed event rate")
    plt.title(f"Calibration: {split} / {task}")
    plt.tight_layout()
    out = OUT_DIR / fname
    plt.savefig(out, dpi=300)
    plt.close()
    print("[OK] saved:", out)

def fig_los_scatter(split: str, model_key: str, disp: str, fname: str):
    df = _load_reg(split, model_key)
    plt.figure()
    plt.scatter(df["y_true"].to_numpy(), df["y_pred"].to_numpy(), s=6)
    plt.xlabel("Observed LOS (days)")
    plt.ylabel("Predicted LOS (days)")
    plt.title(f"Observed vs Predicted LOS: {split} / {disp}")
    plt.tight_layout()
    out = OUT_DIR / fname
    plt.savefig(out, dpi=300)
    plt.close()
    print("[OK] saved:", out)

def fig_los_calibration_by_bins(split: str, model_key: str, disp: str, fname: str, n_bins: int = 10):
    """
    LOS calibration: bin by predicted LOS quantiles -> plot observed mean LOS vs predicted mean LOS.
    """
    df = _load_reg(split, model_key).copy()
    y = df["y_true"].to_numpy(float)
    p = df["y_pred"].to_numpy(float)

    # quantile bins on prediction
    qs = np.linspace(0, 1, n_bins+1)
    cuts = np.quantile(p, qs)
    # dedup cuts
    cuts = np.unique(cuts)
    if len(cuts) <= 2:
        cuts = np.linspace(np.min(p), np.max(p), n_bins+1)

    ids = np.digitize(p, cuts[1:-1], right=True)
    xs, ys = [], []
    for b in range(np.max(ids)+1):
        m = ids == b
        if np.sum(m) == 0:
            continue
        xs.append(float(np.mean(p[m])))
        ys.append(float(np.mean(y[m])))

    plt.figure()
    plt.plot(xs, ys, marker="o")
    # reference
    lo = float(min(min(xs), min(ys)))
    hi = float(max(max(xs), max(ys)))
    plt.plot([lo, hi], [lo, hi])
    plt.xlabel("Mean predicted LOS")
    plt.ylabel("Mean observed LOS")
    plt.title(f"LOS calibration (binned): {split} / {disp}")
    plt.tight_layout()
    out = OUT_DIR / fname
    plt.savefig(out, dpi=300)
    plt.close()
    print("[OK] saved:", out)

def make_all_figures():
    # Calibration plots (classification)
    # models shown: XGB, LGBM, LR_MI (if exists)
    models = []
    for m in CLS_MODELS_TREE:
        models.append(m)
    if INCLUDE_LR_MI:
        models.append("LR_MI")

    for split in ["Holdout", "Temporal"]:
        for task in ["OUTCOME", "LOSBIN"]:
            # skip if no file for task at all
            any_exist = False
            try:
                _ = _load_cls(split, task, CLS_MODELS_TREE[0])
                any_exist = True
            except Exception:
                try:
                    _ = _load_cls(split, task, "LR_MI")
                    any_exist = True
                except Exception:
                    any_exist = False
            if not any_exist:
                continue

            fig_calibration(split, task, models, f"Fig_Calibration_{split}_{task}.png")

        # LOSREG (continuous) — only XGBReg and LGBMRegL1 (display LGBM)
        for key, disp in REG_MODELS:
            fig_los_scatter(split, key, disp, f"Fig_LOS_Scatter_{split}_{disp}.png")
            fig_los_calibration_by_bins(split, key, disp, f"Fig_LOS_CalibBinned_{split}_{disp}.png", n_bins=10)

def run_all():
    if not PRED_DIR.exists():
        raise FileNotFoundError(f"PRED_DIR not found: {PRED_DIR}")

    _inventory()

    # Tables
    df_holdout = make_table_metrics("Holdout")
    df_temporal = make_table_metrics("Temporal")

    # Combined table
    df_all = pd.concat([df_holdout, df_temporal], ignore_index=True)
    df_all.to_excel(OUT_DIR / "Table_metrics_AllSplits.xlsx", index=False)
    print("[OK] saved:", OUT_DIR / "Table_metrics_AllSplits.xlsx")

    # Figures
    make_all_figures()
    print("[DONE] outputs in:", OUT_DIR)

if __name__ == "__main__":
    run_all()
```

```python
# -*- coding: utf-8 -*-
from __future__ import annotations

import re
import gc
import json
import warnings
from dataclasses import dataclass
from pathlib import Path
from typing import Dict, Tuple, List, Optional

import numpy as np
import pandas as pd

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    r2_score,
    mean_squared_error,
    mean_absolute_error,
    median_absolute_error,
)

import statsmodels.api as sm

warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)

# -------------------------
# CONFIG
# -------------------------
RUN_DIR = Path(r"results_paper_ready_seed42")
OUT_DIR = RUN_DIR / "paper_outputs"
PRED_DIR = RUN_DIR / "predictions"

PRED_DIR_CANDIDATES = [
    PRED_DIR,
    OUT_DIR,
    RUN_DIR / "paper_ready_outputs",
    RUN_DIR,
]
INVENTORY_FILE = OUT_DIR / "inventory_predictions.txt"

DPI = 600
SEED = 42
BOOTSTRAP_N = 1000

TASK_OUTCOME = "OUTCOME"
TASK_LOSBIN  = "LOSBIN"
TASK_LOSREG  = "LOSREG"

SPLIT_HOLDOUT  = "Holdout"
SPLIT_TEMPORAL = "Temporal"

MODEL_LR    = "LR"
MODEL_XGB   = "XGB"
MODEL_LGBM  = "LGBM"
MODEL_XGBREG  = "XGBReg"
MODEL_LGBMREG = "LGBMReg"

INDEX_COL_CANDIDATES = ["INDEX", "index", "patient_id", "id", "ID"]

TABLE2_XLSX = OUT_DIR / "Table2_metrics_Holdout_withCI.xlsx"
TABLE3_XLSX = OUT_DIR / "Table3_metrics_Temporal_withCI.xlsx"

# -------------------------
# Utilities
# -------------------------
def _safe_mkdir(p: Path) -> None:
    p.mkdir(parents=True, exist_ok=True)

def _rng(seed: int = SEED) -> np.random.Generator:
    return np.random.default_rng(seed)

def _clip_prob(p: np.ndarray, eps: float = 1e-6) -> np.ndarray:
    p = np.asarray(p, float)
    return np.clip(p, eps, 1.0 - eps)

def _logit(p: np.ndarray, eps: float = 1e-6) -> np.ndarray:
    p = _clip_prob(p, eps=eps)
    return np.log(p / (1.0 - p))

def _smape(y_true: np.ndarray, y_pred: np.ndarray, eps: float = 1e-8) -> float:
    y_true = np.asarray(y_true, float)
    y_pred = np.asarray(y_pred, float)
    denom = (np.abs(y_true) + np.abs(y_pred) + eps)
    return float(np.mean(2.0 * np.abs(y_pred - y_true) / denom))

def _detect_index_col(df: pd.DataFrame) -> Optional[str]:
    for c in INDEX_COL_CANDIDATES:
        if c in df.columns:
            return c
    return None

def _read_pred_csv(path: Path) -> pd.DataFrame:
    df = pd.read_csv(path)
    df.columns = [str(c).strip() for c in df.columns]
    return df

def _infer_pred_cols_classification(df: pd.DataFrame) -> Tuple[str, str]:
    cols = set(df.columns)

    y_candidates = ["y_true", "y", "label", "outcome", "Y"]
    y_col = None
    for c in y_candidates:
        if c in cols:
            y_col = c
            break
    if y_col is None:
        for c in df.columns:
            if c.lower() in ["index", "id", "patient_id"]:
                continue
            s = df[c].dropna()
            if len(s) > 0:
                u = set(pd.unique(s))
                if u.issubset({0, 1, 0.0, 1.0}):
                    y_col = c
                    break
    if y_col is None:
        raise ValueError(f"Cannot find y column in classification predictions: {df.columns.tolist()}")

    p_candidates = ["p_cal", "p", "prob", "proba", "pred_proba", "y_pred", "pred", "prediction"]
    p_col = None
    for c in p_candidates:
        if c in cols:
            p_col = c
            break
    if p_col is None:
        for c in df.columns:
            if c == y_col:
                continue
            if df[c].dtype.kind in "fc":
                s = df[c].dropna()
                if len(s) > 0 and (s.min() >= 0) and (s.max() <= 1):
                    p_col = c
                    break
    if p_col is None:
        raise ValueError(f"Cannot find probability column in classification predictions: {df.columns.tolist()}")

    return y_col, p_col

def _infer_pred_cols_regression(df: pd.DataFrame) -> Tuple[str, str]:
    cols = set(df.columns)

    y_candidates = ["y_true", "y", "label", "outcome", "Y"]
    y_col = None
    for c in y_candidates:
        if c in cols:
            y_col = c
            break
    if y_col is None:
        raise ValueError(f"Cannot find y column in regression predictions: {df.columns.tolist()}")

    pred_candidates = ["y_pred", "pred", "prediction", "yhat", "p", "p_cal"]
    pred_col = None
    for c in pred_candidates:
        if c in cols:
            pred_col = c
            break
    if pred_col is None:
        for c in df.columns:
            if c == y_col:
                continue
            if df[c].dtype.kind in "fc":
                pred_col = c
                break
    if pred_col is None:
        raise ValueError(f"Cannot find prediction column in regression predictions: {df.columns.tolist()}")

    return y_col, pred_col

# -------------------------
# Inventory / Pred map
# -------------------------
@dataclass(frozen=True)
class PredKey:
    split: str
    task: str
    model: str

def _canonicalize_split(s: str) -> Optional[str]:
    s2 = s.lower()
    if "holdout" in s2:
        return SPLIT_HOLDOUT
    if "temporal" in s2:
        return SPLIT_TEMPORAL
    return None

def _canonicalize_task(s: str) -> Optional[str]:
    s2 = s.upper()
    if "OUTCOME" in s2:
        return TASK_OUTCOME
    if "LOSBIN" in s2:
        return TASK_LOSBIN
    if "LOSREG" in s2 or "LOS_REG" in s2 or "LOS-REG" in s2:
        return TASK_LOSREG
    return None

def _canonicalize_model(s: str) -> Optional[str]:
    s2 = s.upper()

    if "LGBM" in s2 and "L1" in s2:
        return None

    # LR includes LR_MIxx
    if "LR_MI" in s2:
        return MODEL_LR
    if "LOGISTIC" in s2 or re.search(r"\bLR\b", s2):
        return MODEL_LR

    if "XGB" in s2 or "XGBOOST" in s2:
        if "REG" in s2:
            return MODEL_XGBREG
        return MODEL_XGB

    if "LGBM" in s2 or "LIGHTGBM" in s2:
        if "REG" in s2:
            return MODEL_LGBMREG
        return MODEL_LGBM

    return None

def _parse_inventory_line(line: str) -> Optional[Path]:
    line = line.strip()
    if not line:
        return None
    p = Path(line)
    if p.exists():
        return p
    p2 = (RUN_DIR / p)
    if p2.exists():
        return p2
    p3 = OUT_DIR / p.name
    if p3.exists():
        return p3
    p4 = PRED_DIR / p.name
    if p4.exists():
        return p4
    return None

def build_pred_map() -> Dict[PredKey, Path]:
    pred_map: Dict[PredKey, Path] = {}
    candidates: List[Path] = []

    if INVENTORY_FILE.exists():
        lines = INVENTORY_FILE.read_text(encoding="utf-8", errors="ignore").splitlines()
        for ln in lines:
            p = _parse_inventory_line(ln)
            if p is not None and p.suffix.lower() in [".csv", ".parquet"]:
                candidates.append(p)

    if not candidates:
        for d in PRED_DIR_CANDIDATES:
            if d.exists():
                candidates.extend(list(d.rglob("*.csv")))

    uniq = []
    seen = set()
    for p in candidates:
        ps = str(p.resolve())
        if ps not in seen:
            uniq.append(p)
            seen.add(ps)
    candidates = uniq

    for p in candidates:
        s = str(p)
        split = _canonicalize_split(s)
        task = _canonicalize_task(s)
        model = _canonicalize_model(s)
        if split is None or task is None or model is None:
            continue

        if task in [TASK_OUTCOME, TASK_LOSBIN] and model in [MODEL_XGBREG, MODEL_LGBMREG]:
            continue
        if task == TASK_LOSREG and model in [MODEL_LR, MODEL_XGB, MODEL_LGBM]:
            continue

        key = PredKey(split=split, task=task, model=model)

        if key not in pred_map:
            pred_map[key] = p
        else:
            old = pred_map[key]
            score_old = 0
            score_new = 0

            if task in [TASK_OUTCOME, TASK_LOSBIN]:
                if "cal" in old.name.lower():
                    score_old += 1
                if "cal" in p.name.lower():
                    score_new += 1

            if model == MODEL_LR:
                if "LR_MI" in old.name.upper():
                    score_old -= 1
                if "LR_MI" in p.name.upper():
                    score_new -= 1

            score_old += -len(old.parts)
            score_new += -len(p.parts)

            if score_new > score_old:
                pred_map[key] = p

    return pred_map

# -------------------------
# LR synthesis from LR_MIxx (required for LR keys)
# -------------------------
def _find_lr_mi_files(split: str, task: str) -> List[Path]:
    task_u = task.upper()
    files: List[Path] = []
    for d in PRED_DIR_CANDIDATES:
        if not d.exists():
            continue
        if split == SPLIT_HOLDOUT:
            files.extend(list(d.rglob(f"{task_u}_LR_MI*_Holdout.csv")))
        elif split == SPLIT_TEMPORAL:
            files.extend(list(d.rglob(f"Temporal_{task_u}_LR_MI*.csv")))
    uniq = sorted({p.resolve() for p in files})
    return [Path(p) for p in uniq]

def _synthesize_lr_from_mi(split: str, task: str, out_path: Path) -> Path:
    mi_files = _find_lr_mi_files(split, task)
    if len(mi_files) == 0:
        raise FileNotFoundError(f"No LR_MI files found to synthesize LR for split={split}, task={task}")

    dfs = []
    for f in mi_files:
        df = _read_pred_csv(f)
        idx_col = _detect_index_col(df)
        if idx_col is None:
            raise ValueError(f"LR_MI file has no INDEX column: {f}")
        y_col, p_col = _infer_pred_cols_classification(df)
        tmp = df[[idx_col, y_col, p_col]].copy()
        tmp.columns = ["INDEX", "y_true", f"p_{f.stem}"]
        dfs.append(tmp)

    base = dfs[0]
    for other in dfs[1:]:
        base = base.merge(other, on=["INDEX", "y_true"], how="inner")

    if base.shape[0] == 0:
        raise RuntimeError(f"After merging LR_MI files, no rows remained (split={split}, task={task}). Check INDEX mismatch.")

    p_cols = [c for c in base.columns if c.startswith("p_")]
    base["p_cal"] = base[p_cols].mean(axis=1).astype(np.float32)

    out_df = base[["INDEX", "y_true", "p_cal"]].copy()
    out_path.parent.mkdir(parents=True, exist_ok=True)
    out_df.to_csv(out_path, index=False, encoding="utf-8-sig")
    print(f"[OK] synthesized LR from {len(mi_files)} MI files -> {out_path}")
    return out_path

def ensure_lr_predictions(pred_map: Dict[PredKey, Path]) -> Dict[PredKey, Path]:
    targets = [
        (SPLIT_HOLDOUT, TASK_OUTCOME),
        (SPLIT_HOLDOUT, TASK_LOSBIN),
        (SPLIT_TEMPORAL, TASK_OUTCOME),
        (SPLIT_TEMPORAL, TASK_LOSBIN),
    ]
    for split, task in targets:
        key = PredKey(split=split, task=task, model=MODEL_LR)
        if key in pred_map and pred_map[key].exists():
            continue
        if split == SPLIT_HOLDOUT:
            out_path = PRED_DIR / f"{task}_LR_Holdout.csv"
        else:
            out_path = PRED_DIR / f"Temporal_{task}_LR.csv"
        pred_map[key] = _synthesize_lr_from_mi(split, task, out_path)
    return pred_map

# -------------------------
# Metrics + Bootstrap
# -------------------------
@dataclass
class ClsMetrics:
    auroc: float
    auprc: float
    brier: float
    cal_slope: float
    citl: float

@dataclass
class RegMetrics:
    rmse: float
    mae: float
    medae: float
    r2: float
    smape: float

def _calibration_slope_citl(y: np.ndarray, p: np.ndarray) -> Tuple[float, float]:
    y = np.asarray(y, float)
    x = _logit(np.asarray(p, float))
    X = sm.add_constant(x, has_constant="add")
    try:
        model = sm.Logit(y, X).fit(disp=0, maxiter=200)
        return float(model.params[1]), float(model.params[0])  # slope, citl
    except Exception:
        return float("nan"), float("nan")

def compute_cls_metrics(y: np.ndarray, p: np.ndarray) -> ClsMetrics:
    y = np.asarray(y, float)
    p = _clip_prob(np.asarray(p, float))
    auroc = float(roc_auc_score(y, p))
    auprc = float(average_precision_score(y, p))
    brier = float(brier_score_loss(y, p))
    slope, citl = _calibration_slope_citl(y, p)
    return ClsMetrics(auroc=auroc, auprc=auprc, brier=brier, cal_slope=slope, citl=citl)

def compute_reg_metrics(y: np.ndarray, pred: np.ndarray) -> RegMetrics:
    y = np.asarray(y, float)
    pred = np.asarray(pred, float)
    rmse = float(np.sqrt(mean_squared_error(y, pred)))
    mae = float(mean_absolute_error(y, pred))
    medae = float(median_absolute_error(y, pred))
    r2 = float(r2_score(y, pred))
    smape = float(_smape(y, pred))
    return RegMetrics(rmse=rmse, mae=mae, medae=medae, r2=r2, smape=smape)

def bootstrap_ci(
    y: np.ndarray,
    x: np.ndarray,
    metric_fn,
    n_boot: int = BOOTSTRAP_N,
    seed: int = SEED,
) -> Tuple[object, Dict[str, Tuple[float, float]]]:
    rng = _rng(seed)
    n = len(y)
    idx = np.arange(n)

    point = metric_fn(y, x)
    fields = [f for f in vars(point).keys()]
    boot_vals = {f: np.zeros(n_boot, dtype=float) for f in fields}

    for b in range(n_boot):
        bs = rng.choice(idx, size=n, replace=True)
        m = metric_fn(y[bs], x[bs])
        for f in fields:
            boot_vals[f][b] = float(getattr(m, f))

    ci = {}
    for f in fields:
        lo, hi = np.nanpercentile(boot_vals[f], [2.5, 97.5])
        ci[f] = (float(lo), float(hi))

    return point, ci

# -------------------------
# Load preds
# -------------------------
def load_cls_predictions(pred_map: Dict[PredKey, Path], split: str, task: str, model: str) -> Tuple[np.ndarray, np.ndarray, Optional[np.ndarray], Optional[pd.DataFrame]]:
    key = PredKey(split=split, task=task, model=model)
    if key not in pred_map:
        raise FileNotFoundError(f"Missing prediction file for {key}")

    df = _read_pred_csv(pred_map[key])
    idx_col = _detect_index_col(df)
    y_col, p_col = _infer_pred_cols_classification(df)

    y = df[y_col].to_numpy()
    p = df[p_col].to_numpy()
    idx = df[idx_col].to_numpy() if idx_col is not None else None
    return y, p, idx, df

def load_reg_predictions(pred_map: Dict[PredKey, Path], split: str, model: str) -> Tuple[np.ndarray, np.ndarray, Optional[np.ndarray], Optional[pd.DataFrame]]:
    key = PredKey(split=split, task=TASK_LOSREG, model=model)
    if key not in pred_map:
        raise FileNotFoundError(f"Missing prediction file for {key}")

    df = _read_pred_csv(pred_map[key])
    idx_col = _detect_index_col(df)
    y_col, pred_col = _infer_pred_cols_regression(df)

    y = df[y_col].to_numpy()
    pred = df[pred_col].to_numpy()
    idx = df[idx_col].to_numpy() if idx_col is not None else None
    return y, pred, idx, df

# -------------------------
# Tables
# -------------------------
def build_tables_with_bootstrap_ci(pred_map: Dict[PredKey, Path]) -> None:
    _safe_mkdir(OUT_DIR)

    def fmt_point_ci(v: float, lo: float, hi: float, digits: int = 3) -> str:
        if not (np.isfinite(v) and np.isfinite(lo) and np.isfinite(hi)):
            return "NA"
        return f"{v:.{digits}f} ({lo:.{digits}f}–{hi:.{digits}f})"

    def build_one_split(split: str) -> pd.DataFrame:
        rows = []

        for task in [TASK_OUTCOME, TASK_LOSBIN]:
            for model in [MODEL_LR, MODEL_XGB, MODEL_LGBM]:
                y, p, _, _ = load_cls_predictions(pred_map, split, task, model)
                y = np.asarray(y).astype(int)
                p = np.asarray(p, float)
                point, ci = bootstrap_ci(y, p, compute_cls_metrics, n_boot=BOOTSTRAP_N, seed=SEED)
                rows.append({
                    "Split": split,
                    "Task": task,
                    "Model": model,
                    "AUROC": fmt_point_ci(point.auroc, *ci["auroc"]),
                    "AUPRC": fmt_point_ci(point.auprc, *ci["auprc"]),
                    "Brier": fmt_point_ci(point.brier, *ci["brier"]),
                    "CalibrationSlope": fmt_point_ci(point.cal_slope, *ci["cal_slope"]),
                    "CITL": fmt_point_ci(point.citl, *ci["citl"]),
                })

        for model in [MODEL_XGBREG, MODEL_LGBMREG]:
            y, pred, _, _ = load_reg_predictions(pred_map, split, model)
            y = np.asarray(y, float)
            pred = np.asarray(pred, float)
            point, ci = bootstrap_ci(y, pred, compute_reg_metrics, n_boot=BOOTSTRAP_N, seed=SEED)
            rows.append({
                "Split": split,
                "Task": TASK_LOSREG,
                "Model": model,
                "RMSE": fmt_point_ci(point.rmse, *ci["rmse"]),
                "MAE": fmt_point_ci(point.mae, *ci["mae"]),
                "MedAE": fmt_point_ci(point.medae, *ci["medae"]),
                "R2": fmt_point_ci(point.r2, *ci["r2"]),
                "SMAPE": fmt_point_ci(point.smape, *ci["smape"]),
            })

        return pd.DataFrame(rows)

    df_holdout  = build_one_split(SPLIT_HOLDOUT)
    df_temporal = build_one_split(SPLIT_TEMPORAL)

    with pd.ExcelWriter(TABLE2_XLSX, engine="openpyxl") as w:
        df_holdout.to_excel(w, index=False, sheet_name="Table2_Holdout")
    print("[OK] saved:", TABLE2_XLSX)

    with pd.ExcelWriter(TABLE3_XLSX, engine="openpyxl") as w:
        df_temporal.to_excel(w, index=False, sheet_name="Table3_Temporal")
    print("[OK] saved:", TABLE3_XLSX)

# -------------------------
# RUN: Table only
# -------------------------
_safe_mkdir(OUT_DIR)
_safe_mkdir(PRED_DIR)

pred_map = build_pred_map()
if not pred_map:
    raise RuntimeError("Prediction map is empty. Ensure prediction CSVs exist under RUN_DIR.")

pred_map = ensure_lr_predictions(pred_map)

# strict required keys for tables
required_for_tables = [
    # Holdout cls
    PredKey(SPLIT_HOLDOUT, TASK_OUTCOME, MODEL_LR),
    PredKey(SPLIT_HOLDOUT, TASK_OUTCOME, MODEL_XGB),
    PredKey(SPLIT_HOLDOUT, TASK_OUTCOME, MODEL_LGBM),
    PredKey(SPLIT_HOLDOUT, TASK_LOSBIN, MODEL_LR),
    PredKey(SPLIT_HOLDOUT, TASK_LOSBIN, MODEL_XGB),
    PredKey(SPLIT_HOLDOUT, TASK_LOSBIN, MODEL_LGBM),
    # Temporal cls
    PredKey(SPLIT_TEMPORAL, TASK_OUTCOME, MODEL_LR),
    PredKey(SPLIT_TEMPORAL, TASK_OUTCOME, MODEL_XGB),
    PredKey(SPLIT_TEMPORAL, TASK_OUTCOME, MODEL_LGBM),
    PredKey(SPLIT_TEMPORAL, TASK_LOSBIN, MODEL_LR),
    PredKey(SPLIT_TEMPORAL, TASK_LOSBIN, MODEL_XGB),
    PredKey(SPLIT_TEMPORAL, TASK_LOSBIN, MODEL_LGBM),
    # Holdout reg
    PredKey(SPLIT_HOLDOUT, TASK_LOSREG, MODEL_XGBREG),
    PredKey(SPLIT_HOLDOUT, TASK_LOSREG, MODEL_LGBMREG),
    # Temporal reg
    PredKey(SPLIT_TEMPORAL, TASK_LOSREG, MODEL_XGBREG),
    PredKey(SPLIT_TEMPORAL, TASK_LOSREG, MODEL_LGBMREG),
]
missing = [k for k in required_for_tables if (k not in pred_map or (not pred_map[k].exists()))]
if missing:
    raise FileNotFoundError("Missing required prediction files for TABLES:\n" + "\n".join([f"- {m}" for m in missing]))

build_tables_with_bootstrap_ci(pred_map)

print("[DONE] Table outputs in:", OUT_DIR)
```

```python
# -*- coding: utf-8 -*-
from __future__ import annotations

from pathlib import Path
import numpy as np
import pandas as pd

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image

from sklearn.metrics import roc_curve, precision_recall_curve

# -------------------------------------------------
# 前提（セル1で定義済み）：
# pred_map, load_cls_predictions, load_reg_predictions,
# _safe_mkdir, OUT_DIR, DPI,
# TASK_OUTCOME, TASK_LOSBIN,
# SPLIT_HOLDOUT,
# MODEL_LR, MODEL_XGB, MODEL_LGBM,
# MODEL_XGBREG, MODEL_LGBMREG
# -------------------------------------------------

# ===== 最終出力は JPEG =====
FIG2_JPG  = OUT_DIR / "Figure2_4panel_AUROC_Calibration_OUTCOME_LOSBIN.jpg"
FIG3_JPG  = OUT_DIR / "Figure3_2panel_DCA_OUTCOME_LOSBIN.jpg"
FIGS1_JPG = OUT_DIR / "FigS1_2panel_AUPRC_OUTCOME_LOSBIN.jpg"
FIGS2_JPG = OUT_DIR / "FigS2_2panel_LOSREG_Scatter_Overall.jpg"
FIGS3_JPG = OUT_DIR / "FigS3_4panel_LOSREG_Scatter_ByOutcomeEvent.jpg"

TITLE_FONTSIZE = 14
XLIM_LOS = 60.0

# =================================================
# 共通ヘルパ
# =================================================
def _set_bold_title(ax: plt.Axes, title: str, fontsize: int = TITLE_FONTSIZE) -> None:
    """
    mathtext($\\bf{...}$)内でスペースが潰れないように、
    通常の半角スペースを自動で '\\ ' に変換する。
    → 呼び出し側は普通の文字列を書けばよい。
    """
    safe = title.replace("\\", r"\\").replace(" ", r"\ ")
    ax.set_title(r"$\bf{" + safe + "}$", fontsize=fontsize)

def _save_panel(fig: plt.Figure, out_png: Path) -> None:
    out_png.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(out_png, dpi=DPI, bbox_inches="tight")
    plt.close(fig)

def _merge_pngs_grid(
    png_paths,
    out_jpg: Path,
    nrows: int,
    ncols: int,
    pad: int = 25,
    quality: int = 95
) -> None:
    imgs = [Image.open(p).convert("RGB") for p in png_paths]
    w = max(i.width for i in imgs)
    h = max(i.height for i in imgs)

    canvas = Image.new(
        "RGB",
        (ncols * w + (ncols + 1) * pad, nrows * h + (nrows + 1) * pad),
        (255, 255, 255)
    )

    for i, im in enumerate(imgs):
        r, c = divmod(i, ncols)
        x = pad + c * (w + pad) + (w - im.width) // 2
        y = pad + r * (h + pad) + (h - im.height) // 2
        canvas.paste(im, (x, y))

    out_jpg.parent.mkdir(parents=True, exist_ok=True)
    canvas.save(out_jpg, format="JPEG", quality=quality, subsampling=0)
    for im in imgs:
        im.close()

    print("[OK] saved (JPEG):", out_jpg)

# =================================================
# Calibration bins
# =================================================
def _calibration_bins(y_true: np.ndarray, p: np.ndarray, n_bins: int = 10) -> pd.DataFrame:
    y = np.asarray(y_true, float)
    p = np.clip(np.asarray(p, float), 1e-6, 1 - 1e-6)

    qs = np.linspace(0, 1, n_bins + 1)
    edges = np.quantile(p, qs)
    edges[0] = 0.0
    edges[-1] = 1.0
    for i in range(1, len(edges)):
        if edges[i] <= edges[i - 1]:
            edges[i] = min(1.0, edges[i - 1] + 1e-6)

    bin_id = np.digitize(p, edges[1:-1], right=True)
    df = pd.DataFrame({"y": y, "p": p, "bin": bin_id})
    g = df.groupby("bin", as_index=False).agg(
        mean_pred=("p", "mean"),
        mean_obs=("y", "mean"),
        n=("y", "size"),
    )
    return g

# =================================================
# Figure2: ROC / Calibration（凡例＝モデル名のみ）
# =================================================
def panel_roc(split: str, task: str, tag: str, out_png: Path) -> None:
    fig = plt.figure(figsize=(7.0, 6.0))
    ax = fig.add_subplot(111)

    _set_bold_title(ax, f"{tag} {task} AUROC ({split})")
    ax.set_xlabel("False positive rate")
    ax.set_ylabel("True positive rate")

    for model, name in [(MODEL_LR, "LR"), (MODEL_XGB, "XGBoost"), (MODEL_LGBM, "LightGBM")]:
        y, p, _, _ = load_cls_predictions(pred_map, split, task, model)
        fpr, tpr, _ = roc_curve(np.asarray(y).astype(int), np.asarray(p, float))
        ax.plot(fpr, tpr, linewidth=1.5, label=name)

    ax.plot([0, 1], [0, 1], linewidth=1, linestyle="--")
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.legend(loc="lower right", fontsize=9)
    _save_panel(fig, out_png)

def panel_calibration(split: str, task: str, tag: str, out_png: Path, n_bins: int = 10) -> None:
    fig = plt.figure(figsize=(7.0, 6.0))
    ax = fig.add_subplot(111)

    _set_bold_title(ax, f"{tag} {task} Calibration ({split})")
    ax.set_xlabel("Mean predicted probability (bin)")
    ax.set_ylabel("Observed event rate (bin)")
    ax.plot([0, 1], [0, 1], linewidth=1, linestyle="--")

    for model, name in [(MODEL_LR, "LR"), (MODEL_XGB, "XGBoost"), (MODEL_LGBM, "LightGBM")]:
        y, p, _, _ = load_cls_predictions(pred_map, split, task, model)
        g = _calibration_bins(np.asarray(y).astype(int), np.asarray(p, float), n_bins)
        ax.plot(g["mean_pred"], g["mean_obs"], marker="o", linewidth=1.5, label=name)

    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.legend(loc="lower right", fontsize=9)
    _save_panel(fig, out_png)

# =================================================
# FigS1: PR curves
# =================================================
def panel_pr(split: str, task: str, tag: str, out_png: Path) -> None:
    fig = plt.figure(figsize=(7.0, 6.0))
    ax = fig.add_subplot(111)

    _set_bold_title(ax, f"{tag} {task} AUPRC ({split})")
    ax.set_xlabel("Recall")
    ax.set_ylabel("Precision")

    for model, name in [(MODEL_LR, "LR"), (MODEL_XGB, "XGBoost"), (MODEL_LGBM, "LightGBM")]:
        y, p, _, _ = load_cls_predictions(pred_map, split, task, model)
        prec, rec, _ = precision_recall_curve(np.asarray(y).astype(int), np.asarray(p, float))
        ax.plot(rec, prec, linewidth=1.5, label=name)

    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.legend(loc="lower left", fontsize=9)
    _save_panel(fig, out_png)

# =================================================
# Figure3: DCA（縦軸 -0.6~0.6）
# =================================================
def _net_benefit(y: np.ndarray, p: np.ndarray, pt: float) -> float:
    y = np.asarray(y).astype(int)
    p = np.asarray(p, float)
    treat = (p >= pt).astype(int)
    tp = int(np.sum((treat == 1) & (y == 1)))
    fp = int(np.sum((treat == 1) & (y == 0)))
    n = len(y)
    w = pt / (1.0 - pt)
    return (tp / n) - (fp / n) * w

def panel_dca(split: str, task: str, tag: str, out_png: Path,
              pt_min: float = 0.01, pt_max: float = 0.70, n_pts: int = 200) -> None:
    fig = plt.figure(figsize=(7.0, 6.0))
    ax = fig.add_subplot(111)

    _set_bold_title(ax, f"{tag} {task} Decision Curve Analysis ({split})")
    ax.set_xlabel("Threshold probability")
    ax.set_ylabel("Net benefit")

    y0, _, _, _ = load_cls_predictions(pred_map, split, task, MODEL_LR)
    prev = float(np.mean(np.asarray(y0).astype(int)))
    pts = np.linspace(pt_min, pt_max, n_pts)

    ax.plot(pts, np.zeros_like(pts), linewidth=1, linestyle="--", label="Treat-none")
    ax.plot(pts, np.array([prev - (1 - prev) * (pt / (1 - pt)) for pt in pts]),
            linewidth=1, linestyle="--", label="Treat-all")

    for model, name in [(MODEL_LR, "LR"), (MODEL_XGB, "XGBoost"), (MODEL_LGBM, "LightGBM")]:
        y, p, _, _ = load_cls_predictions(pred_map, split, task, model)
        nb = np.array([_net_benefit(y, p, pt) for pt in pts])
        ax.plot(pts, nb, linewidth=1.5, label=name)

    ax.axhline(0.0, linewidth=1)
    ax.set_xlim(pt_min, pt_max)
    ax.set_ylim(-0.6, 0.6)
    ax.legend(loc="best", fontsize=9)
    _save_panel(fig, out_png)

# =================================================
# FigS2: LOS scatter（Overall, x<=60）
# =================================================
def panel_los_scatter_overall(split: str, model: str, tag: str, out_png: Path) -> None:
    y, pred, _, _ = load_reg_predictions(pred_map, split, model)
    fig = plt.figure(figsize=(7.0, 6.0))
    ax = fig.add_subplot(111)

    name = "XGBoost" if model == MODEL_XGBREG else "LightGBM"
    _set_bold_title(ax, f"{tag} LOS Scatter (Overall) - {name}")
    ax.set_xlabel("Predicted LOS")
    ax.set_ylabel("Observed LOS")

    ax.scatter(pred, y, s=8, alpha=0.25)
    ax.set_xlim(0, XLIM_LOS)
    ax.plot([0, XLIM_LOS], [0, XLIM_LOS], linewidth=1)
    _save_panel(fig, out_png)

# =================================================
# FigS3: LOS scatter（Home / Non-home で層別）
# =================================================
def build_figS3_stratified(split: str) -> None:
    y_out, _, idx_out, _ = load_cls_predictions(pred_map, split, TASK_OUTCOME, MODEL_LR)
    out_df = pd.DataFrame({"INDEX": idx_out, "OUTCOME": np.asarray(y_out).astype(int)})

    yx, px, ix, _ = load_reg_predictions(pred_map, split, MODEL_XGBREG)
    yl, pl, il, _ = load_reg_predictions(pred_map, split, MODEL_LGBMREG)

    df_x = pd.DataFrame({"INDEX": ix, "y_true": yx, "y_pred": px}).merge(out_df, on="INDEX")
    df_l = pd.DataFrame({"INDEX": il, "y_true": yl, "y_pred": pl}).merge(out_df, on="INDEX")

    panel_dir = OUT_DIR / "_panels_figS3"; _safe_mkdir(panel_dir)

    def _scatter(df, outcome, tag, title, out_png):
        sub = df[df["OUTCOME"] == outcome]
        fig = plt.figure(figsize=(7.0, 6.0))
        ax = fig.add_subplot(111)
        _set_bold_title(ax, f"{tag} {title}")
        ax.set_xlabel("Predicted LOS")
        ax.set_ylabel("Observed LOS")
        ax.scatter(sub["y_pred"], sub["y_true"], s=8, alpha=0.25)
        ax.set_xlim(0, XLIM_LOS)
        ax.plot([0, XLIM_LOS], [0, XLIM_LOS], linewidth=1)
        _save_panel(fig, out_png)

    pA = panel_dir / "FigS3_A_Home_XGBReg.png"
    pB = panel_dir / "FigS3_B_Home_LGBMReg.png"
    pC = panel_dir / "FigS3_C_Nonhome_XGBReg.png"
    pD = panel_dir / "FigS3_D_Nonhome_LGBMReg.png"

    _scatter(df_x, 0, "(A)", "LOS Scatter (Home) - XGBoost", pA)
    _scatter(df_l, 0, "(B)", "LOS Scatter (Home) - LightGBM", pB)
    _scatter(df_x, 1, "(C)", "LOS Scatter (Non-home) - XGBoost", pC)
    _scatter(df_l, 1, "(D)", "LOS Scatter (Non-home) - LightGBM", pD)

    _merge_pngs_grid([pA, pB, pC, pD], FIGS3_JPG, 2, 2)

# =================================================
# RUN
# =================================================
def build_all_nonshap():
    s = SPLIT_HOLDOUT

    d2 = OUT_DIR / "_panels_fig2"; _safe_mkdir(d2)
    pA = d2 / "Fig2_A.png"; pB = d2 / "Fig2_B.png"; pC = d2 / "Fig2_C.png"; pD = d2 / "Fig2_D.png"
    panel_roc(s, TASK_OUTCOME, "(A)", pA)
    panel_calibration(s, TASK_OUTCOME, "(B)", pB)
    panel_roc(s, TASK_LOSBIN, "(C)", pC)
    panel_calibration(s, TASK_LOSBIN, "(D)", pD)
    _merge_pngs_grid([pA, pB, pC, pD], FIG2_JPG, 2, 2)

    d3 = OUT_DIR / "_panels_fig3"; _safe_mkdir(d3)
    pA3 = d3 / "Fig3_A.png"; pB3 = d3 / "Fig3_B.png"
    panel_dca(s, TASK_OUTCOME, "(A)", pA3)
    panel_dca(s, TASK_LOSBIN, "(B)", pB3)
    _merge_pngs_grid([pA3, pB3], FIG3_JPG, 1, 2)

    ds1 = OUT_DIR / "_panels_figS1"; _safe_mkdir(ds1)
    ps1A = ds1 / "FigS1_A.png"; ps1B = ds1 / "FigS1_B.png"
    panel_pr(s, TASK_OUTCOME, "(A)", ps1A)
    panel_pr(s, TASK_LOSBIN, "(B)", ps1B)
    _merge_pngs_grid([ps1A, ps1B], FIGS1_JPG, 1, 2)

    ds2 = OUT_DIR / "_panels_figS2"; _safe_mkdir(ds2)
    ps2A = ds2 / "FigS2_A.png"; ps2B = ds2 / "FigS2_B.png"
    panel_los_scatter_overall(s, MODEL_XGBREG, "(A)", ps2A)
    panel_los_scatter_overall(s, MODEL_LGBMREG, "(B)", ps2B)
    _merge_pngs_grid([ps2A, ps2B], FIGS2_JPG, 1, 2)

    build_figS3_stratified(s)
    print("[DONE] non-SHAP figures (JPEG) in:", OUT_DIR)

build_all_nonshap()
```

```python
# =========================
# SHAP (ONE CELL): Figure4(XGB) + FigureS4(LR_MI) + FigureS5(LGBM)
# - RAWは読まない（shap_data/ と models/ を直接使う）
# - 出力は JPEG
# - (A)/(B) はタイトル先頭
# - タイトル文字は少し大きめ
# =========================
from __future__ import annotations

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import shap
from PIL import Image
import joblib

# =========================
# CONFIG
# =========================
RANDOM_STATE = 42
RUN_DIR = Path(f"results_paper_ready_seed{RANDOM_STATE}")
MODEL_DIR = RUN_DIR / "models"
SHAP_DIR  = RUN_DIR / "shap_data"
OUT_DIR   = RUN_DIR / "figures_shap"; OUT_DIR.mkdir(exist_ok=True)

# Split: "Holdout" or "Temporal"
SPLIT = "Holdout"

# tasks
TASK_A = "OUTCOME"
TASK_B = "LOSBIN"

# output files
FIG4_JPG = OUT_DIR / f"Figure4_SHAP_{SPLIT}_XGB.jpg"      # XGB
FIGS4_JPG = OUT_DIR / f"FigureS4_SHAP_{SPLIT}_LR_MI.jpg"  # LR + MI (MI01 by default)
FIGS5_JPG = OUT_DIR / f"FigureS5_SHAP_{SPLIT}_LGBM.jpg"   # LGBM

# LR+MI: which MI to use for SHAP (存在するものを指定)
LR_MI_TAG = "LR_MI01"   # "LR_MI01" ... "LR_MI10" etc

# plot controls
MAX_DISPLAY = 20
MAX_N = 3000
TITLE_FONTSIZE = 16
PNG_DPI = 300
MERGE_PAD = 35

# =========================
# UTIL
# =========================
def _load_shap_X(split: str, task: str, model_tag: str) -> pd.DataFrame:
    """
    shap_data の X を読む
      X_{split}_{task}_{model_tag}.parquet or .csv
    model_tag: "XGB" / "LGBM" / "LR_MI01" etc
    """
    p_parq = SHAP_DIR / f"X_{split}_{task}_{model_tag}.parquet"
    p_csv  = SHAP_DIR / f"X_{split}_{task}_{model_tag}.csv"
    if p_parq.exists():
        df = pd.read_parquet(p_parq)
    elif p_csv.exists():
        df = pd.read_csv(p_csv)
    else:
        raise FileNotFoundError(f"SHAP X not found: {p_parq} or {p_csv}")
    df.columns = [str(c).strip() for c in df.columns]
    return df

def _load_model_extras(split: str, task: str, model_tag: str) -> dict:
    """
    models の joblib を読む
    Holdout:
      {task}_{model_tag}_Holdout_final.joblib
      e.g., OUTCOME_XGB_Holdout_final.joblib / OUTCOME_LR_MI01_Holdout_final.joblib
    Temporal:
      Temporal_{task}_{model_tag}_final.joblib
      e.g., Temporal_OUTCOME_XGB_final.joblib / Temporal_OUTCOME_LR_MI01_final.joblib
    """
    if split == "Holdout":
        p = MODEL_DIR / f"{task}_{model_tag}_Holdout_final.joblib"
    elif split == "Temporal":
        p = MODEL_DIR / f"Temporal_{task}_{model_tag}_final.joblib"
    else:
        raise ValueError("SPLIT must be 'Holdout' or 'Temporal'")
    if not p.exists():
        raise FileNotFoundError(f"Model extras joblib not found: {p}")
    return joblib.load(p)

def _is_tree_model(m) -> bool:
    name = m.__class__.__name__.lower()
    if "xgb" in name or "lgbm" in name or "lightgbm" in name:
        return True
    if hasattr(m, "get_booster"):
        return True
    return False

def _downsample_df(X: pd.DataFrame, max_n: int | None, seed: int = 42) -> pd.DataFrame:
    if max_n is None or len(X) <= max_n:
        return X.reset_index(drop=True)
    rng = np.random.default_rng(seed)
    idx = rng.choice(np.arange(len(X)), size=max_n, replace=False)
    return X.iloc[idx].reset_index(drop=True)

def _merge_pngs_to_jpg(png_paths: list[Path], out_jpg: Path, pad: int = 35, bg=(255,255,255)) -> None:
    imgs = [Image.open(str(p)).convert("RGB") for p in png_paths]
    widths = [im.width for im in imgs]
    heights = [im.height for im in imgs]
    total_w = sum(widths) + pad * (len(imgs) + 1)
    max_h = max(heights) + pad * 2
    canvas = Image.new("RGB", (total_w, max_h), bg)
    x = pad
    for im in imgs:
        y = (max_h - im.height) // 2
        canvas.paste(im, (x, y))
        x += im.width + pad
    canvas.save(str(out_jpg), quality=95)

def _save_shap_summary_png(
    X_feat: pd.DataFrame,
    model,
    out_png: Path,
    title: str,
    max_display: int = 20,
    max_n: int = 3000,
    title_fontsize: int = 16,
) -> None:
    """
    SHAP summary dot plot -> PNG
    """
    X_use = _downsample_df(X_feat, max_n=max_n, seed=42)

    # explainer
    if _is_tree_model(model):
        explainer = shap.TreeExplainer(model)
        shap_values = explainer.shap_values(X_use)
        if isinstance(shap_values, list):
            shap_values = shap_values[-1]  # binary: positive class
        values_for_plot = shap_values
        X_for_plot = X_use
        is_linear = False
    else:
        # Linear model (LR)
        # NOTE: model is a fitted LogisticRegression; we explain in feature space used for training.
        masker = shap.maskers.Independent(X_use)
        explainer = shap.LinearExplainer(model, masker=masker)
        exp = explainer(X_use)
        values_for_plot = exp
        X_for_plot = X_use
        is_linear = True

    # plot
    plt.figure(figsize=(8.5, 6.0))
    shap.summary_plot(
        values_for_plot,
        X_for_plot,
        plot_type="dot",
        max_display=max_display,
        show=False
    )
    plt.title(title, fontsize=title_fontsize)
    plt.tight_layout()
    plt.savefig(out_png, dpi=PNG_DPI, bbox_inches="tight")
    plt.close()

# =========================
# RUN HELPERS
# =========================
def _build_two_panel_jpg(
    model_tag: str,
    out_jpg: Path,
    title_A: str,
    title_B: str,
    task_A: str = "OUTCOME",
    task_B: str = "LOSBIN",
) -> None:
    panel_dir = OUT_DIR / f"panels_{model_tag}_{SPLIT}"
    panel_dir.mkdir(exist_ok=True)

    # ---- A ----
    xA = _load_shap_X(SPLIT, task_A, model_tag)
    XA = xA.drop(columns=["INDEX"], errors="ignore")
    extrasA = _load_model_extras(SPLIT, task_A, model_tag)
    modelA = extrasA["final_model"]

    pA = panel_dir / f"A_{SPLIT}_{model_tag}_{task_A}.png"
    _save_shap_summary_png(
        X_feat=XA,
        model=modelA,
        out_png=pA,
        title=title_A,
        max_display=MAX_DISPLAY,
        max_n=MAX_N,
        title_fontsize=TITLE_FONTSIZE
    )

    # ---- B ----
    xB = _load_shap_X(SPLIT, task_B, model_tag)
    XB = xB.drop(columns=["INDEX"], errors="ignore")
    extrasB = _load_model_extras(SPLIT, task_B, model_tag)
    modelB = extrasB["final_model"]

    pB = panel_dir / f"B_{SPLIT}_{model_tag}_{task_B}.png"
    _save_shap_summary_png(
        X_feat=XB,
        model=modelB,
        out_png=pB,
        title=title_B,
        max_display=MAX_DISPLAY,
        max_n=MAX_N,
        title_fontsize=TITLE_FONTSIZE
    )

    _merge_pngs_to_jpg([pA, pB], out_jpg, pad=MERGE_PAD)

# =========================
# 1) Figure 4: XGB
# =========================
_build_two_panel_jpg(
    model_tag="XGB",
    out_jpg=FIG4_JPG,
    title_A="(A) SHAP summary plot  OUTCOME  (XGB)",
    title_B="(B) SHAP summary plot  LOSBIN  (XGB)",
    task_A=TASK_A,
    task_B=TASK_B,
)
print("[DONE] Figure4:", FIG4_JPG)

# =========================
# 2) Supplementary Figure S4: LR + MI (MI01 by default)
# =========================
_build_two_panel_jpg(
    model_tag=LR_MI_TAG,
    out_jpg=FIGS4_JPG,
    title_A=f"(A) SHAP summary plot  OUTCOME  ({LR_MI_TAG})",
    title_B=f"(B) SHAP summary plot  LOSBIN  ({LR_MI_TAG})",
    task_A=TASK_A,
    task_B=TASK_B,
)
print("[DONE] FigureS4:", FIGS4_JPG)

# =========================
# 3) Supplementary Figure S5: LGBM
# =========================
_build_two_panel_jpg(
    model_tag="LGBM",
    out_jpg=FIGS5_JPG,
    title_A="(A) SHAP summary plot  OUTCOME  (LGBM)",
    title_B="(B) SHAP summary plot  LOSBIN  (LGBM)",
    task_A=TASK_A,
    task_B=TASK_B,
)
print("[DONE] FigureS5:", FIGS5_JPG)

print("\n[ALL DONE] Outputs in:", OUT_DIR.resolve())
```

```python
# -*- coding: utf-8 -*-
"""
Sensitivity analyses (Holdout only) + Paper-ready Table 2 (with 95% bootstrap CI)
- Sens1: exclude TARGET_OUTCOME == 2
- Sens2: LOSBIN cutoff = 28
- Output: one Excel containing BOTH sensitivity scenarios in Table 2 format
Assumptions:
- The main pipeline code cell has already been executed in this notebook, and defines:
  main_holdout, load_and_build_table, RUN_DIR, RANDOM_STATE, MODE, _cfg,
  TARGET_OUTCOME, LOS_CUTOFF_DAYS, EVAL_XLSX, (optionally) PRED_DIR, OUT_DIR, FIG_DIR, etc.
- The pipeline writes prediction CSVs somewhere based on RUN_DIR and/or PRED_DIR.
If it currently does NOT write prediction CSVs, CI cannot be computed (this script will error with a clear message).
"""

from __future__ import annotations

import copy
import math
import re
from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.metrics import roc_auc_score, average_precision_score
from sklearn.linear_model import LogisticRegression

# =========================================================
# 0) Guard: pipeline globals exist
# =========================================================
_required = [
    "main_holdout",
    "load_and_build_table",
    "RUN_DIR",
    "RANDOM_STATE",
    "MODE",
    "_cfg",
    "TARGET_OUTCOME",
    "LOS_CUTOFF_DAYS",
    "EVAL_XLSX",
]
missing = [k for k in _required if k not in globals()]
if missing:
    raise RuntimeError(
        "Pipeline本体がこのNotebookで未定義です。先に本体（巨大コード）セルを実行してください。\n"
        f"Missing: {missing}"
    )

# =========================================================
# 1) Config
# =========================================================
SEED = int(globals()["RANDOM_STATE"])
BASE_DIR = Path(".").resolve()

SENS1_SCENARIO = "Sens1_DropEvent2"
SENS2_SCENARIO = "Sens2_LOS28"

# Holdout only (as requested)
RUN_WHICH = "holdout"

# Bootstrap settings (adjust if heavy)
N_BOOT = 2000
BOOT_SEED = 20250109

# Output Excel
OUT_XLSX = BASE_DIR / f"Table2_Sensitivity_Holdout_seed{SEED}.xlsx"

# Display labels (paper)
SENS1_LABEL = "Sensitivity 1 (Outcome=2 excluded)"
SENS2_LABEL = "Sensitivity 2 (LOSBIN cutoff=28)"

# LOSBIN label (paper) for each scenario
# Sens1 keeps the original LOS cutoff from the main pipeline
LOS_LABEL_SENS1 = f"Length of stay ≤{int(globals()['LOS_CUTOFF_DAYS'])} days (LOSBIN)"
LOS_LABEL_SENS2 = "Length of stay ≤28 days (LOSBIN)"

# =========================================================
# 2) Helpers: safely overwrite path-like globals used by pipeline
# =========================================================
def _set_if_exists(g: dict, key: str, value):
    if key in g:
        g[key] = value

def _backup_keys(g: dict, keys: list[str]) -> dict:
    bk = {}
    for k in keys:
        if k in g:
            bk[k] = copy.deepcopy(g[k])
    return bk

def _restore_keys(g: dict, backup: dict):
    for k, v in backup.items():
        g[k] = v

# =========================================================
# 3) Sensitivity runner (Holdout only) — FIXED: update PRED_DIR/OUT_DIR too
# =========================================================
def run_sensitivity_holdout(
    scenario: str,
    *,
    drop_event2: bool,
    los_cutoff: int | None,
) -> Path:
    """
    Run holdout for a sensitivity scenario by overwriting BOTH RUN_DIR and (if present) derived dirs like PRED_DIR/OUT_DIR.
    Returns: run_dir (scenario folder)
    """
    g = globals()

    # keys we may need to overwrite (depends on your pipeline)
    path_keys = [
        "RUN_DIR",
        "PRED_DIR",
        "OUT_DIR",
        "FIG_DIR",
        "EVAL_XLSX",
        "EVAL_TEMP_XLSX",
    ]
    scalar_keys = [
        "RANDOM_STATE",
        "MODE",
        "LOS_CUTOFF_DAYS",
        "_cfg",
        "load_and_build_table",
    ]

    backup = {}
    backup.update(_backup_keys(g, path_keys))
    backup.update(_backup_keys(g, scalar_keys))

    try:
        # --- LOS cutoff override ---
        if los_cutoff is not None:
            g["LOS_CUTOFF_DAYS"] = int(los_cutoff)

        # --- scenario run_dir ---
        run_dir = BASE_DIR / f"results_paper_ready_seed{int(g['RANDOM_STATE'])}__{scenario}"
        run_dir.mkdir(parents=True, exist_ok=True)

        # --- overwrite run dir & common output paths ---
        g["RUN_DIR"] = run_dir
        _set_if_exists(g, "EVAL_XLSX", run_dir / "Evaluation_All.xlsx")
        _set_if_exists(g, "EVAL_TEMP_XLSX", run_dir / "Evaluation_Temporal.xlsx")

        # --- overwrite prediction/output dirs if pipeline uses them ---
        _set_if_exists(g, "PRED_DIR", run_dir / "predictions")
        _set_if_exists(g, "OUT_DIR", run_dir / "paper_outputs")
        _set_if_exists(g, "FIG_DIR", run_dir / "figures")

        # ensure dirs exist if keys exist
        if "PRED_DIR" in g:
            Path(g["PRED_DIR"]).mkdir(parents=True, exist_ok=True)
        if "OUT_DIR" in g:
            Path(g["OUT_DIR"]).mkdir(parents=True, exist_ok=True)
        if "FIG_DIR" in g:
            Path(g["FIG_DIR"]).mkdir(parents=True, exist_ok=True)

        # --- Sens1: drop outcome==2 by monkeypatching loader ---
        if drop_event2:
            orig_loader = backup["load_and_build_table"]

            def load_and_build_table_drop2():
                tbl, feat_cols = orig_loader()
                before = tbl.shape[0]
                tbl = tbl.loc[tbl[g["TARGET_OUTCOME"]] != 2].reset_index(drop=True)
                after = tbl.shape[0]
                return tbl, feat_cols

            g["load_and_build_table"] = load_and_build_table_drop2

        # --- run HOLDOUT only ---
        g["main_holdout"]()

        return run_dir

    finally:
        _restore_keys(g, backup)

# =========================================================
# 4) Locate prediction CSV directory (robust)
# =========================================================
def locate_pred_dir(run_dir: Path) -> Path:
    """
    Accepts:
      - run_dir/predictions/*.csv  (preferred)
      - or any *.csv under run_dir that matches expected naming patterns
    Returns directory that contains the prediction csv files.
    """
    pred = run_dir / "predictions"
    if pred.exists() and any(pred.glob("*.csv")):
        return pred

    # fallback: search recursively for a directory that contains expected CSV patterns
    # expected examples (from your earlier script):
    # OUTCOME_XGB_Holdout.csv, OUTCOME_LGBM_Holdout.csv, OUTCOME_LR_MI01_Holdout.csv ...
    # LOSBIN_XGB_Holdout.csv, LOSREG_XGBReg_Holdout.csv etc.
    pat = re.compile(r"^(OUTCOME|LOSBIN|LOSREG)_.+_Holdout\.csv$", re.IGNORECASE)

    candidates = {}
    for p in run_dir.rglob("*.csv"):
        if pat.match(p.name):
            candidates.setdefault(p.parent, 0)
            candidates[p.parent] += 1

    if candidates:
        # choose directory with the most matches
        best = sorted(candidates.items(), key=lambda x: x[1], reverse=True)[0][0]
        return best

    raise FileNotFoundError(
        f"No prediction CSVs found under: {run_dir}\n"
        f"Expected: (OUTCOME|LOSBIN|LOSREG)_*_Holdout.csv\n"
        f"Hint: pipelineが予測CSVを書き出す設定（RUN_DIR/PRED_DIR連動）を確認してください。"
    )

# =========================================================
# 5) Load prediction CSVs (Holdout)
# =========================================================
def _read_csv(path: Path) -> pd.DataFrame:
    return pd.read_csv(path)

def _find_mi_files(pred_dir: Path, prefix: str) -> list[Path]:
    # prefix example: "OUTCOME_LR_MI" -> OUTCOME_LR_MI01_Holdout.csv ...
    pat = re.compile(rf"^{re.escape(prefix)}(\d{{2}})_Holdout\.csv$", re.IGNORECASE)
    hits = []
    for p in pred_dir.glob("*.csv"):
        m = pat.match(p.name)
        if m:
            hits.append((int(m.group(1)), p))
    hits = [p for _, p in sorted(hits, key=lambda x: x[0])]
    return hits

def load_cls_holdout(pred_dir: Path, task: str, model: str) -> pd.DataFrame:
    """
    returns columns: INDEX, y_true, p_cal
    model: "LR" (MI平均), "XGB", "LGBM"
    """
    task = task.upper()

    if model in ("XGB", "LGBM"):
        f = pred_dir / f"{task}_{model}_Holdout.csv"
        if not f.exists():
            raise FileNotFoundError(f"Missing: {f}")
        df = _read_csv(f)
        need = {"INDEX", "y_true", "p_cal"}
        if not need.issubset(df.columns):
            raise ValueError(f"{f.name} columns={list(df.columns)} required={sorted(need)}")
        return df[["INDEX", "y_true", "p_cal"]].copy()

    if model == "LR":
        files = _find_mi_files(pred_dir, f"{task}_LR_MI")
        if len(files) == 0:
            # allow non-MI single file naming if exists
            f_single = pred_dir / f"{task}_LR_Holdout.csv"
            if f_single.exists():
                df = _read_csv(f_single)
                need = {"INDEX", "y_true", "p_cal"}
                if not need.issubset(df.columns):
                    raise ValueError(f"{f_single.name} columns={list(df.columns)} required={sorted(need)}")
                return df[["INDEX", "y_true", "p_cal"]].copy()
            raise FileNotFoundError(f"No LR MI files for {task} (expected {task}_LR_MI01_Holdout.csv ...)")
        dfs = [_read_csv(p) for p in files]
        for d, p in zip(dfs, files):
            need = {"INDEX", "y_true", "p_cal"}
            if not need.issubset(d.columns):
                raise ValueError(f"{p.name} columns={list(d.columns)} required={sorted(need)}")
        base = dfs[0][["INDEX", "y_true"]].copy()
        ps = [d.set_index("INDEX")["p_cal"] for d in dfs]
        pmat = pd.concat(ps, axis=1).loc[base["INDEX"]].to_numpy(float)
        out = base.copy()
        out["p_cal"] = np.mean(pmat, axis=1).astype(np.float64)
        return out

    raise ValueError("model must be 'LR','XGB','LGBM'")

def load_reg_holdout(pred_dir: Path, model_key: str) -> pd.DataFrame:
    """
    returns columns: INDEX, y_true, y_pred
    model_key: "XGBReg" or "LGBMRegL1" or "LGBMReg"
    """
    # tolerate a few variants
    candidates = [
        pred_dir / f"LOSREG_{model_key}_Holdout.csv",
        pred_dir / "LOSREG_XGBReg_Holdout.csv" if model_key == "XGBReg" else None,
    ]
    candidates = [c for c in candidates if c is not None]

    f = None
    for c in candidates:
        if c.exists():
            f = c
            break
    if f is None:
        raise FileNotFoundError(f"Missing LOSREG holdout CSV for {model_key} in {pred_dir}")

    df = _read_csv(f)
    need = {"INDEX", "y_true", "y_pred"}
    if not need.issubset(df.columns):
        raise ValueError(f"{f.name} columns={list(df.columns)} required={sorted(need)}")
    return df[["INDEX", "y_true", "y_pred"]].copy()

# =========================================================
# 6) Metrics + calibration slope/CITL
# =========================================================
def _clip_p(p: np.ndarray) -> np.ndarray:
    p = np.asarray(p, float)
    return np.clip(p, 1e-6, 1 - 1e-6)

def _logit(p: np.ndarray) -> np.ndarray:
    p = _clip_p(p)
    return np.log(p / (1 - p))

def cls_point_metrics(y: np.ndarray, p: np.ndarray) -> dict:
    y = np.asarray(y, int)
    p = _clip_p(p)
    out = {}
    # discrimination
    out["AUROC"] = float(roc_auc_score(y, p)) if np.unique(y).size == 2 else np.nan
    out["AUPRC"] = float(average_precision_score(y, p)) if np.unique(y).size == 2 else np.nan
    # brier
    out["Brier"] = float(np.mean((p - y) ** 2))

    # calibration slope: logistic regression y ~ logit(p)
    x = _logit(p).reshape(-1, 1)
    try:
        lr = LogisticRegression(penalty=None, solver="lbfgs", max_iter=2000)
        lr.fit(x, y)
        out["CalSlope"] = float(lr.coef_.ravel()[0])
    except Exception:
        out["CalSlope"] = np.nan

    # CITL: intercept-only logistic with offset=logit(p)
    # Implement as: fit logistic y ~ 1 with fixed offset, via Newton iteration on intercept.
    # This avoids statsmodels dependency.
    o = _logit(p)
    b = 0.0
    for _ in range(50):
        eta = b + o
        mu = 1 / (1 + np.exp(-eta))
        w = mu * (1 - mu)
        # prevent division by zero
        w = np.clip(w, 1e-12, None)
        z = eta + (y - mu) / w
        # weighted least squares update for intercept only
        b_new = np.sum(w * (z - o)) / np.sum(w)
        if abs(b_new - b) < 1e-10:
            b = b_new
            break
        b = b_new
    out["CITL"] = float(b)

    return out

def reg_point_metrics(y: np.ndarray, yhat: np.ndarray) -> dict:
    y = np.asarray(y, float)
    yhat = np.asarray(yhat, float)
    rmse = float(np.sqrt(np.mean((y - yhat) ** 2)))
    mae = float(np.mean(np.abs(y - yhat)))
    medae = float(np.median(np.abs(y - yhat)))
    ss_res = float(np.sum((y - yhat) ** 2))
    ss_tot = float(np.sum((y - np.mean(y)) ** 2))
    r2 = float(1 - ss_res / ss_tot) if ss_tot > 0 else np.nan
    denom = np.abs(y) + np.abs(yhat)
    denom = np.where(denom == 0, 1.0, denom)
    smape = float(np.mean(2 * np.abs(yhat - y) / denom))
    return {"RMSE": rmse, "MAE": mae, "MedAE": medae, "R2": r2, "SMAPE": smape}

def bootstrap_ci(
    y: np.ndarray,
    p_or_yhat: np.ndarray,
    metric_fn,
    metric_keys: list[str],
    n_boot: int,
    seed: int,
) -> dict:
    """
    Percentile bootstrap CI (2.5, 97.5).
    metric_fn returns dict of metrics.
    """
    rng = np.random.default_rng(seed)
    n = len(y)
    stats = {k: [] for k in metric_keys}

    for b in range(n_boot):
        idx = rng.integers(0, n, size=n)
        yy = y[idx]
        pp = p_or_yhat[idx]
        m = metric_fn(yy, pp)
        for k in metric_keys:
            v = m.get(k, np.nan)
            stats[k].append(v)

    out = {}
    for k in metric_keys:
        arr = np.asarray(stats[k], float)
        arr = arr[np.isfinite(arr)]
        if arr.size == 0:
            out[k] = (np.nan, np.nan, np.nan)
            continue
        lo, hi = np.quantile(arr, [0.025, 0.975])
        out[k] = (float(lo), float(hi), float(np.mean(arr)))
    return out

def fmt_point_ci(point: float, lo: float, hi: float, digits: int = 3) -> str:
    if not np.isfinite(point) or not np.isfinite(lo) or not np.isfinite(hi):
        return ""
    return f"{point:.{digits}f} ({lo:.{digits}f}–{hi:.{digits}f})"

# =========================================================
# 7) Build Table 2 blocks (Sensitivity 1 + 2), HOLDOUT only
# =========================================================
def build_table2_for_scenario(
    pred_dir: Path,
    scenario_label: str,
    losbin_label: str,
    *,
    drop_outcome2: bool,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Returns:
      cls_block_df, reg_block_df
    Table format matches your "Table 2" example:
      A. Classification tasks ... rows per task x model
      B. Regression task ... rows per model
    """
    # --- classification tasks: OUTCOME, LOSBIN ---
    # Display names must match paper:
    model_map = [("LR", "Logistic regression"), ("XGB", "XGBoost"), ("LGBM", "LightGBM")]

    cls_rows = []
    for task, task_label in [
        ("OUTCOME", "Discharge outcome (home vs non-home)"),
        ("LOSBIN", losbin_label),
    ]:
        for mkey, mdisp in model_map:
            df = load_cls_holdout(pred_dir, task, mkey)

            # Sens1 requires dropping outcome==2 (only relevant for OUTCOME task; but safe to apply only there)
            if drop_outcome2 and task == "OUTCOME":
                df = df.loc[df["y_true"].astype(int) != 2].copy()
                # convert to binary if your pipeline encoded home/non-home differently
                # Here we assume y_true is already binary 0/1 in OUTCOME after your pipeline,
                # but when outcome==2 exists, it indicates a 3-class label. After removing 2, keep 0/1.
                df["y_true"] = df["y_true"].astype(int)

            y = df["y_true"].to_numpy(int)
            p = df["p_cal"].to_numpy(float)

            point = cls_point_metrics(y, p)

            ci = bootstrap_ci(
                y, p,
                metric_fn=cls_point_metrics,
                metric_keys=["AUROC", "AUPRC", "Brier", "CalSlope", "CITL"],
                n_boot=N_BOOT,
                seed=BOOT_SEED,
            )
            # Use bootstrap mean of metric as "center" can drift; we want point estimate from full sample
            # so we format using point + percentile CI
            row = {
                "Scenario": scenario_label,
                "A. Classification tasks (OUTCOME and LOSBIN)": task_label,
                "Model": mdisp,
                "AUROC (95% CI)": fmt_point_ci(point["AUROC"], ci["AUROC"][0], ci["AUROC"][1], digits=3),
                "AUPRC (95% CI)": fmt_point_ci(point["AUPRC"], ci["AUPRC"][0], ci["AUPRC"][1], digits=3),
                "Brier score (95% CI)": fmt_point_ci(point["Brier"], ci["Brier"][0], ci["Brier"][1], digits=3),
                "Calibration slope (95% CI)": fmt_point_ci(point["CalSlope"], ci["CalSlope"][0], ci["CalSlope"][1], digits=3),
                "CITL (95% CI)": fmt_point_ci(point["CITL"], ci["CITL"][0], ci["CITL"][1], digits=3),
            }
            cls_rows.append(row)

    cls_df = pd.DataFrame(cls_rows)

    # --- regression task: LOSREG ---
    reg_rows = []
    for key, disp in [("XGBReg", "XGBReg"), ("LGBMRegL1", "LGBMReg")]:
        df = load_reg_holdout(pred_dir, key)
        y = df["y_true"].to_numpy(float)
        yhat = df["y_pred"].to_numpy(float)

        point = reg_point_metrics(y, yhat)
        ci = bootstrap_ci(
            y, yhat,
            metric_fn=reg_point_metrics,
            metric_keys=["RMSE", "MAE", "MedAE", "R2", "SMAPE"],
            n_boot=N_BOOT,
            seed=BOOT_SEED,
        )

        reg_rows.append({
            "Scenario": scenario_label,
            "B. Regression task (LOSREG)": " ",
            "Model": disp,
            "RMSE (95% CI)": fmt_point_ci(point["RMSE"], ci["RMSE"][0], ci["RMSE"][1], digits=3),
            "MAE (95% CI)": fmt_point_ci(point["MAE"], ci["MAE"][0], ci["MAE"][1], digits=3),
            "Median AE (95% CI)": fmt_point_ci(point["MedAE"], ci["MedAE"][0], ci["MedAE"][1], digits=3),
            "R² (95% CI)": fmt_point_ci(point["R2"], ci["R2"][0], ci["R2"][1], digits=3),
            "SMAPE (95% CI)": fmt_point_ci(point["SMAPE"], ci["SMAPE"][0], ci["SMAPE"][1], digits=3),
        })

    reg_df = pd.DataFrame(reg_rows)
    return cls_df, reg_df

# =========================================================
# 8) Run: Sens1 + Sens2 (holdout only) -> build Table2 -> write Excel
# =========================================================
def write_excel(cls_df: pd.DataFrame, reg_df: pd.DataFrame, out_xlsx: Path):
    with pd.ExcelWriter(out_xlsx, engine="openpyxl") as w:
        # keep the same “Table 2” feeling: a single sheet is often preferred
        table2 = pd.concat([cls_df, reg_df], ignore_index=True)
        table2.to_excel(w, sheet_name="Table2_Sensitivity_Holdout", index=False)

        # optional: separate blocks
        cls_df.to_excel(w, sheet_name="A_Classification", index=False)
        reg_df.to_excel(w, sheet_name="B_Regression", index=False)

def run():
    # --- Run sensitivity analyses (holdout only) ---
    dir1 = run_sensitivity_holdout(
        SENS1_SCENARIO,
        drop_event2=True,
        los_cutoff=None,
    )
    dir2 = run_sensitivity_holdout(
        SENS2_SCENARIO,
        drop_event2=False,
        los_cutoff=28,
    )

    # --- Locate prediction CSVs ---
    pred1 = locate_pred_dir(dir1)
    pred2 = locate_pred_dir(dir2)

    print("[INFO] Sens1 run_dir:", dir1)
    print("[INFO] Sens1 pred_dir:", pred1)
    print("[INFO] Sens2 run_dir:", dir2)
    print("[INFO] Sens2 pred_dir:", pred2)

    # --- Build paper-ready Table 2 (Holdout only) ---
    cls1, reg1 = build_table2_for_scenario(
        pred1,
        SENS1_LABEL,
        LOS_LABEL_SENS1,
        drop_outcome2=True,
    )
    cls2, reg2 = build_table2_for_scenario(
        pred2,
        SENS2_LABEL,
        LOS_LABEL_SENS2,
        drop_outcome2=False,
    )

    cls_all = pd.concat([cls1, cls2], ignore_index=True)
    reg_all = pd.concat([reg1, reg2], ignore_index=True)

    # --- Write Excel ---
    write_excel(cls_all, reg_all, OUT_XLSX)
    print("[OK] saved:", OUT_XLSX.resolve())

run()
```
